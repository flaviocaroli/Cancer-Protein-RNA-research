### The following runs in 64 mins locally

```python
# Set up paths
current_dir = os.getcwd()          # or os.path.dirname(os.path.abspath("new_correlations.ipynb"))
data_dir = os.path.join(current_dir, "data")
csv_correlations_dir = os.path.join(data_dir, "correlations")
joblib_correlations_dir = os.path.join(data_dir, "correlations_joblib")

# Create the new folder for joblib files
os.makedirs(joblib_correlations_dir, exist_ok=True)

# Convert each CSV file to joblib
for fname in os.listdir(csv_correlations_dir):
    if fname.endswith(".csv"):
        csv_path = os.path.join(csv_correlations_dir, fname)
        
        # Read the CSV (usually 1 row per file, but can be more)
        df = pd.read_csv(csv_path, index_col=0)
        
        # Construct output filename (replace .csv with .joblib)
        outname = fname.replace(".csv", ".joblib")
        outpath = os.path.join(joblib_correlations_dir, outname)
        
        # Save as joblib
        dump(df, outpath)
        print(f"Converted {fname} -> {outname}")
```

### Pearson

merge all the joblib files

```python
joblib_correlations_dir = os.path.join(data_dir, "correlations_joblib")

# Final master output paths
master_joblib_file = os.path.join(data_dir, "full_pearson.joblib")
master_csv_file = os.path.join(data_dir, "full_pearson.csv")

# Gather all joblib files
all_joblib_files = [
    f for f in os.listdir(joblib_correlations_dir) 
    if f.endswith(".joblib")
]

if not all_joblib_files:
    print(f"No .joblib files found in {joblib_correlations_dir}")
else:
    dfs = []
    for fname in all_joblib_files:
        # Derive the protein name by removing the trailing "_corr.joblib"
        # e.g., "A1BG_corr.joblib" -> "A1BG"
        protein_name = re.sub(r"_corr\.joblib$", "", fname)

        fpath = os.path.join(joblib_correlations_dir, fname)
        df_or_series = load(fpath)  # Load the DataFrame or Series

        # If it's a Series, convert it to a one-row DataFrame
        if isinstance(df_or_series, pd.Series):
            df = df_or_series.to_frame().T
        else:
            df = df_or_series

        # Force the row index to be the protein name
        # (assuming there's only 1 row in each file)
        df.index = [protein_name]

        dfs.append(df)
    
    # Merge all DataFrames into one
    master_df = pd.concat(dfs, axis=0)
    master_df.index.name = "proteins"
    master_df = master_df.astype("float32")
    # Save the merged DataFrame as a joblib file
    dump(master_df, master_joblib_file)

    print(f"Master correlation matrix saved as joblib to: {master_joblib_file}")
```
