# Raw, Z-score, and Linear Proxy / ElasticNet / XGBoost plots

This notebook merges the three original plotting notebooks into one deduplicated workflow.

It loads the transcriptomics/proteomics data and saved ENet/XGBoost H5 summaries once, recovers the exact saved H5 test split once, and then builds three same-gene proxy variants:

1. **Raw Proxy** — test RNA values are used directly.
2. **Z-score Proxy** — test RNA is standardized with training RNA statistics and mapped to the training protein scale.
3. **Linear Proxy** — a train-only `LinearRegression` maps same-gene RNA to protein abundance.

The shared metric-selection, summary, plotting, and export functions are defined only once. The notebook produces both a five-model overview and the three original proxy-vs-ENet-vs-XGBoost comparisons.

Run this notebook from the first cell to the final export cell using the original data files. Fonts, subplot spacing, title wrapping, and statistics boxes have been enlarged.
```python
# >>> pRNA figure routing (added by the pRNA update; keep as the first code cell)
# Every figure is shown inline. Every file written with savefig (png, pdf, svg) is redirected to
# figures_pdf/<notebook name>/ next to this notebook; sub-folders of the original path are preserved.
import os as _fig_os
from pathlib import Path as _FigPath
import matplotlib as _fig_mpl
import matplotlib.pyplot as _fig_plt
from matplotlib.figure import Figure as _FigFigure

NOTEBOOK_NAME = "0_proxies_enet_xgb_plots"
FIG_DIR = (_FigPath.cwd() / "figures_pdf" / NOTEBOOK_NAME).resolve()
FIG_DIR.mkdir(parents=True, exist_ok=True)


def _route_figure_path(fname):
    p = _FigPath(_fig_os.fspath(fname))
    full = p if p.is_absolute() else (_FigPath.cwd() / p)
    try:
        full.resolve().relative_to(FIG_DIR)
        return fname                                  # already inside this notebook's figure folder
    except ValueError:
        pass
    try:
        rel = _FigPath(_fig_os.path.relpath(full.resolve(), _FigPath.cwd()))
    except ValueError:                                # e.g. another drive on Windows
        rel = _FigPath(p.name)
    parts = [x for x in rel.parts if x not in ("..", ".", "figures_pdf", NOTEBOOK_NAME)]
    target = FIG_DIR.joinpath(*parts) if parts else FIG_DIR / p.name
    target.parent.mkdir(parents=True, exist_ok=True)
    return target


if not getattr(_FigFigure.savefig, "_prna_routed", False):
    _fig_orig_savefig = _FigFigure.savefig

    def _fig_routed_savefig(self, fname, *args, **kwargs):
        if isinstance(fname, (str, bytes, _FigPath)) or hasattr(fname, "__fspath__"):
            fname = _route_figure_path(fname)
        return _fig_orig_savefig(self, fname, *args, **kwargs)

    _fig_routed_savefig._prna_routed = True
    _FigFigure.savefig = _fig_routed_savefig

    _fig_orig_close = _fig_plt.close

    def _fig_close_and_show(fig=None):
        """Display figures that are closed before Jupyter could render them, then close."""
        import inspect as _fig_inspect
        _caller = _fig_inspect.currentframe().f_back
        _caller_mod = (_caller.f_globals.get("__name__", "") if _caller is not None else "")
        if _caller_mod.startswith(("matplotlib", "ipykernel")):   # backend housekeeping, not user code
            return _fig_orig_close(fig)
        if "inline" in str(_fig_mpl.get_backend()).lower():
            from IPython.display import display as _fig_display
            open_nums = set(_fig_plt.get_fignums())
            if fig is None:
                figs = [_fig_plt.gcf()] if open_nums else []
            elif isinstance(fig, str) and fig == "all":
                figs = [_fig_plt.figure(n) for n in sorted(open_nums)]
            elif isinstance(fig, _FigFigure):
                figs = [fig] if getattr(fig, "number", None) in open_nums else []
            elif isinstance(fig, int):
                figs = [_fig_plt.figure(fig)] if fig in open_nums else []
            else:
                figs = []
            for _f in figs:
                if not getattr(_f, "_prna_displayed", False):
                    _f._prna_displayed = True
                    _fig_display(_f)
        return _fig_orig_close(fig)

    _fig_plt.close = _fig_close_and_show

    _fig_orig_show = _fig_plt.show

    def _fig_show_and_mark(*args, **kwargs):
        for _n in _fig_plt.get_fignums():                    # shown now -> never re-displayed on close
            _fig_plt.figure(_n)._prna_displayed = True
        return _fig_orig_show(*args, **kwargs)

    _fig_plt.show = _fig_show_and_mark

print("Figures are saved to", FIG_DIR)
```

```python
# Larger plot text and deferred PDF export.
# Keep this cell before the analysis/plotting cells.
from pathlib import Path
from collections import OrderedDict
from io import BytesIO
import os
import re
import textwrap
import warnings
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib._pylab_helpers import Gcf
from matplotlib.ticker import MaxNLocator

NOTEBOOK_SLUG = 'proxies_enet_xgb'
REMOVE_PLOT_TITLES = True
NOTEBOOK_DIRECTORY_OVERRIDE = None  # Set a Path here only if the kernel runs elsewhere.


def _notebook_directory():
    if NOTEBOOK_DIRECTORY_OVERRIDE is not None:
        return Path(NOTEBOOK_DIRECTORY_OVERRIDE).expanduser().resolve()
    # VS Code and Jupyter may provide the notebook's actual path.
    candidates = [globals().get('__vsc_ipynb_file__'), os.environ.get('JPY_SESSION_NAME')]
    for candidate in candidates:
        if candidate:
            path = Path(candidate).expanduser()
            if path.suffix == '.ipynb' and path.is_file():
                return path.resolve().parent
    try:
        import ipynbname
        return ipynbname.path().resolve().parent
    except (ImportError, FileNotFoundError, RuntimeError, OSError):
        # Standard Jupyter kernels start in the notebook's own directory.
        return Path.cwd().resolve()


NOTEBOOK_DIRECTORY = _notebook_directory()
FIGURES_PDF_DIR = FIG_DIR
_pdf_figure_cache = OrderedDict()
_pdf_next_number = 1

mpl.rcParams.update({
    'font.size': 13,
    'axes.titlesize': 17,
    'axes.labelsize': 14,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 12,
    'legend.title_fontsize': 12,
    'figure.titlesize': 22,
    'figure.figsize': (8, 6.5),
    'pdf.fonttype': 42,
    'ps.fonttype': 42,
    'savefig.bbox': 'tight',
    'savefig.pad_inches': 0.25,
})


def set_readable_theme(*args, **kwargs):
    """Keep Seaborn's own implicit label/tick fonts larger as well."""
    import seaborn as sns
    kwargs.setdefault('font_scale', 1.20)
    return sns.set_theme(*args, **kwargs)


def readable_figsize(original_size):
    """Increase physical space together with fonts, including dynamic grids."""
    width, height = original_size
    if NOTEBOOK_SLUG == 'hpa_barplots':
        return float(width) * 1.18, float(height) * 1.35
    return float(width) * 1.10, float(height) * 1.25


def _wrap_to_width(artist, renderer, max_width):
    """Wrap whole words using measured text extents, without shortening labels."""
    original = getattr(artist, '_readable_original_text', artist.get_text())
    artist._readable_original_text = original
    if not original or '$' in original:
        return
    artist.set_text(original)
    if artist.get_window_extent(renderer).width <= max_width:
        return
    for width in range(max(12, max(map(len, original.splitlines())) - 1), 11, -1):
        candidate = '\n'.join(
            textwrap.fill(line, width=width, break_long_words=False, break_on_hyphens=False)
            for line in original.split('\n')
        )
        artist.set_text(candidate)
        if artist.get_window_extent(renderer).width <= max_width:
            break



def _remember_figure_name(fig, text):
    """Keep descriptive export names without displaying a plot title."""
    fig._readable_export_title = str(text)


def _remember_axes_name(ax, text):
    if not hasattr(ax.figure, '_readable_export_axes_title'):
        ax.figure._readable_export_axes_title = str(text)


def prepare_readable_figure(fig):
    """Apply minimum font sizes and measure margins before display or export."""
    if not fig.axes or getattr(fig, '_readable_preparing', False):
        return fig
    fig._readable_preparing = True
    try:
        if REMOVE_PLOT_TITLES:
            for axis in fig.axes:
                for location in ('left', 'center', 'right'):
                    title_artist = axis.set_title('', loc=location)
                    title_artist._readable_original_text = ''
            if fig._suptitle is not None:
                fig._suptitle.remove()
                fig._suptitle = None
        if getattr(fig, '_compact_protein_metrics', False):
            # Short, wide layout for stacking five panels on an A4 page.
            fig.tight_layout(pad=0.25, w_pad=0.8, h_pad=0.25)
            fig.canvas.draw()
            return fig
        for ax in fig.axes:
            if not ax.get_visible():
                continue
            ax.title.set_fontsize(max(16, ax.title.get_fontsize()))
            ax.xaxis.label.set_fontsize(max(14, ax.xaxis.label.get_fontsize()))
            ax.yaxis.label.set_fontsize(max(14, ax.yaxis.label.get_fontsize()))
            ax.xaxis.labelpad = max(8, ax.xaxis.labelpad)
            ax.yaxis.labelpad = max(9, ax.yaxis.labelpad)
            ax.tick_params(axis='both', which='major', pad=5)
            for label in ax.get_xticklabels() + ax.get_yticklabels():
                label.set_fontsize(max(11.5, label.get_fontsize()))
            for offset in (ax.xaxis.get_offset_text(), ax.yaxis.get_offset_text()):
                offset.set_fontsize(max(11.5, offset.get_fontsize()))
            for artist in ax.texts:
                artist.set_fontsize(max(11.5, artist.get_fontsize()))
                if re.fullmatch(r'\(?[A-Za-z]\)', artist.get_text().strip()):
                    # Panel letters sit outside the data and statistics boxes.
                    artist.set_position((-0.06, 1.03))
                    artist.set_ha('right')
                    artist.set_va('bottom')
                    artist.set_clip_on(False)
                if not ax.axison and artist.get_text().count('\n') >= 8:
                    # Keep the full statistics box within its own panel.
                    artist.set_position((0.03, 0.96))
                    artist.set_ha('left')
                    artist.set_va('top')
                    artist.set_linespacing(1.10)
            legend = ax.get_legend()
            if legend is not None:
                for label in legend.get_texts():
                    label.set_fontsize(max(11.5, label.get_fontsize()))
                legend.get_title().set_fontsize(max(12, legend.get_title().get_fontsize()))
        if fig._suptitle is not None:
            fig._suptitle.set_fontsize(max(20, fig._suptitle.get_fontsize()))
            fig._suptitle.set_y(0.985)
            fig._suptitle.set_va('top')
        for legend in fig.legends:
            for label in legend.get_texts():
                label.set_fontsize(max(11.5, label.get_fontsize()))
        # Repeat after layout: wrapping changes title height and required margins.
        for _ in range(3):
            fig.canvas.draw()
            renderer = fig.canvas.get_renderer()
            if fig._suptitle is not None:
                _wrap_to_width(fig._suptitle, renderer, fig.bbox.width * 0.94)
            for ax in fig.axes:
                if ax.get_visible():
                    _wrap_to_width(ax.title, renderer, ax.bbox.width * 0.96)
            bottom, top = 0.02, 0.98
            if fig._suptitle is not None:
                title_height = fig._suptitle.get_window_extent(renderer).height / fig.bbox.height
                top = min(top, 0.97 - title_height - 0.018)
            for legend in fig.legends:
                bounds = legend.get_window_extent(renderer)
                fraction = bounds.height / fig.bbox.height
                if bounds.y0 + bounds.height / 2 < fig.bbox.y0 + fig.bbox.height / 2:
                    bottom = max(bottom, fraction + 0.045)
                else:
                    top = min(top, 0.97 - fraction - 0.025)
            # Let tight_layout control GridSpec spacing instead of skipping grids
            # whose hspace/wspace was explicitly fixed by older plotting code.
            engine = fig.get_layout_engine()
            if engine is None or engine.__class__.__name__ != 'ConstrainedLayoutEngine':
                grids = set()
                for ax in fig.axes:
                    if hasattr(ax, 'get_subplotspec'):
                        spec = ax.get_subplotspec()
                        if spec is not None:
                            grids.add(spec.get_gridspec())
                for grid in grids:
                    grid.update(hspace=None, wspace=None)
            if engine is not None and engine.__class__.__name__ == 'ConstrainedLayoutEngine':
                engine.set(w_pad=0.12, h_pad=0.16, wspace=0.10, hspace=0.15)
            else:
                with warnings.catch_warnings():
                    warnings.filterwarnings('ignore', message='.*not compatible with tight_layout.*')
                    fig.tight_layout(pad=1.25, h_pad=2.1, w_pad=2.2, rect=(0.01, bottom, 0.99, top))
            fig.canvas.draw()
            renderer = fig.canvas.get_renderer()
            # A long statistics block must not extend below its allocated panel.
            required_growth = 1.0
            for ax in fig.axes:
                for artist in ax.texts:
                    # Reserve room inside axes for bar-end numbers and log-scale
                    # percentage annotations; do not move their data coordinates.
                    if artist.get_transform() is ax.transData and ax.axison:
                        bounds = artist.get_window_extent(renderer)
                        x0, x1 = ax.get_xlim()
                        y0, y1 = ax.get_ylim()
                        if x1 > x0 and bounds.x1 > ax.bbox.x1 + 1:
                            right = ax.transData.inverted().transform((bounds.x1 + 8, ax.bbox.y0))[0]
                            if np.isfinite(right):
                                ax.set_xlim(x0, right)
                        if y1 > y0 and bounds.y1 > ax.bbox.y1 + 1:
                            upper = ax.transData.inverted().transform((ax.bbox.x0, bounds.y1 + 8))[1]
                            if np.isfinite(upper):
                                ax.set_ylim(y0, upper)
                    if not ax.axison and artist.get_text().count('\n') >= 8:
                        bounds = artist.get_window_extent(renderer)
                        required_growth = max(required_growth, bounds.height / max(ax.bbox.height * 0.94, 1))
                # Numeric ticks can use fewer positions while preserving their scale.
                labels = [t for t in ax.get_xticklabels() if t.get_visible() and t.get_text()]
                boxes = [t.get_window_extent(renderer) for t in labels]
                overlaps = any(a.overlaps(b) for a, b in zip(boxes, boxes[1:]))
                if overlaps and not ax.xaxis.get_units() and len(labels) > 5:
                    if all(t.get_rotation() == 0 for t in labels):
                        locator = ax.xaxis.get_major_locator()
                        if locator.__class__.__name__ in {'AutoLocator', 'MaxNLocator'}:
                            ax.xaxis.set_major_locator(MaxNLocator(nbins=4))
            if required_growth > 1.01:
                width, height = fig.get_size_inches()
                fig.set_size_inches(width, height * min(required_growth * 1.05, 1.6), forward=True)
        # Avoid collisions between annotation text and an axes legend.
        fig.canvas.draw()
        renderer = fig.canvas.get_renderer()
        for ax in fig.axes:
            legend = ax.get_legend()
            if legend is None:
                continue
            legend_box = legend.get_window_extent(renderer)
            if any(t.get_visible() and t.get_text() and legend_box.overlaps(t.get_window_extent(renderer))
                   for t in ax.texts):
                if hasattr(legend, 'set_loc'):
                    for location in ('upper left', 'lower right', 'lower left'):
                        legend.set_loc(location)
                        fig.canvas.draw()
                        legend_box = legend.get_window_extent(renderer)
                        if not any(t.get_visible() and t.get_text() and legend_box.overlaps(t.get_window_extent(renderer))
                                   for t in ax.texts):
                            break
        fig.canvas.draw()
        return fig
    finally:
        fig._readable_preparing = False


def _cache_figure_pdf(fig, source_name=None):
    """Snapshot vector PDFs in memory so closed figures survive until the last cell."""
    global _pdf_next_number
    if not fig.axes or not any(ax.get_visible() for ax in fig.axes):
        return
    key = getattr(fig, '_readable_pdf_key', None)
    if key is None or not str(key).startswith(NOTEBOOK_SLUG + '__'):
        key = f'{NOTEBOOK_SLUG}__{_pdf_next_number:03d}'
        _pdf_next_number += 1
        fig._readable_pdf_key = key
    previous = _pdf_figure_cache.get(key, {})
    name = previous.get('name')
    if source_name is not None and isinstance(source_name, (str, os.PathLike)):
        name = Path(source_name).stem
    if not name:
        title = getattr(fig, '_readable_export_title', '')
        if not title:
            title = getattr(fig, '_readable_export_axes_title', '')
        if not title:
            title = fig._suptitle.get_text() if fig._suptitle is not None else ''
        if not title:
            title = next((ax.get_title() for ax in fig.axes if ax.get_title()), 'figure')
        name = title
    name = re.sub(r'[^A-Za-z0-9._-]+', '_', name).strip('._')[:120] or 'figure'
    buffer = BytesIO()
    fig.savefig(buffer, format='pdf', bbox_inches='tight',
                pad_inches=getattr(fig, '_readable_pdf_pad', 0.25),
                metadata={'Creator': 'Matplotlib', 'Title': name})
    _pdf_figure_cache[key] = {
        'name': name, 'pdf': buffer.getvalue(),
        'export_name': getattr(fig, '_readable_export_name', None),
    }


def save_readable_figure(fig, filename, *args, **kwargs):
    prepare_readable_figure(fig)
    kwargs.setdefault('bbox_inches', 'tight')
    kwargs.setdefault('pad_inches', 0.25)
    result = fig.savefig(filename, *args, **kwargs)
    _cache_figure_pdf(fig, source_name=filename)
    return result


def save_readable_current_figure(filename, *args, **kwargs):
    return save_readable_figure(plt.gcf(), filename, *args, **kwargs)


def show_readable_figures(*args, **kwargs):
    for manager in list(Gcf.get_all_fig_managers()):
        fig = manager.canvas.figure
        prepare_readable_figure(fig)
        _cache_figure_pdf(fig)
    return plt.show(*args, **kwargs)


def close_readable_figures(fig=None):
    if fig == 'all':
        figures = [m.canvas.figure for m in Gcf.get_all_fig_managers()]
    elif fig is None:
        manager = Gcf.get_active()
        figures = [manager.canvas.figure] if manager is not None else []
    elif hasattr(fig, 'savefig'):
        figures = [fig]
    else:
        manager = Gcf.get_fig_manager(fig) if isinstance(fig, int) else None
        figures = [manager.canvas.figure] if manager is not None else []
    for figure in figures:
        prepare_readable_figure(figure)
        _cache_figure_pdf(figure)
    return plt.close(fig)


def export_notebook_figures():
    """Write every figure created during this run, including already-closed figures."""
    for manager in list(Gcf.get_all_fig_managers()):
        fig = manager.canvas.figure
        prepare_readable_figure(fig)
        _cache_figure_pdf(fig)
    if not _pdf_figure_cache:
        raise RuntimeError('No figures captured. Run the setup and plotting cells first.')
    FIGURES_PDF_DIR.mkdir(parents=True, exist_ok=True)
    saved = []
    for key, record in _pdf_figure_cache.items():
        filename = record.get('export_name') or f'{key}__{record["name"]}.pdf'
        path = FIGURES_PDF_DIR / filename
        try:
            path.write_bytes(record['pdf'])
        except PermissionError:
            # Save a new copy if the existing PDF is locked.
            from uuid import uuid4

            path = path.with_name(
                f"{path.stem}__{uuid4().hex[:8]}{path.suffix}"
            )
            path.write_bytes(record['pdf'])
    saved.append(path)
    print(f'Saved {len(saved)} PDF figures to: {FIGURES_PDF_DIR}')
    for path in saved:
        print('  ' + path.name)
    return saved

print('Readable plotting enabled. Final PDF destination:', FIGURES_PDF_DIR)
```

```python
# ---------------------------------------------------------------------
# Fallback helpers used only if utils_ccle did not provide them.
# ---------------------------------------------------------------------

def _define_if_missing(name, obj):
    if name not in globals():
        globals()[name] = obj


def _fallback_require_file(path):
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"Required file not found: {path.resolve()}")
    return path


def _fallback_split_hash(samples):
    values = [str(x) for x in pd.Index(samples)]
    payload = "\n".join(values).encode("utf-8")
    return hashlib.sha1(payload).hexdigest()[:12]


def _as_python_scalar(x):
    if isinstance(x, bytes):
        return x.decode("utf-8", errors="replace")
    if isinstance(x, np.generic):
        return x.item()
    if isinstance(x, np.ndarray):
        if x.shape == ():
            return _as_python_scalar(x.item())
        if x.size == 1:
            return _as_python_scalar(x.reshape(-1)[0])
        return [_as_python_scalar(v) for v in x.reshape(-1).tolist()]
    return x


def _decode_array(values):
    arr = np.asarray(values)
    if arr.shape == ():
        return pd.Index([_as_python_scalar(arr.item())]).astype(str)
    return pd.Index([_as_python_scalar(v) for v in arr.reshape(-1)]).astype(str)


def _h5_visit_items(h5):
    items = []
    def visitor(name, obj):
        items.append((name, obj))
    h5.visititems(visitor)
    return items


def _h5_find_attr(h5, candidate_names):
    wanted = {c.lower() for c in candidate_names}
    groups = [("/", h5)] + [(name, obj) for name, obj in _h5_visit_items(h5) if hasattr(obj, "attrs")]
    for _, obj in groups:
        for key, value in obj.attrs.items():
            if str(key).lower() in wanted:
                return _as_python_scalar(value)
    return None


def _h5_find_dataset(h5, candidate_names):
    wanted = {c.lower() for c in candidate_names}
    for name, obj in _h5_visit_items(h5):
        if isinstance(obj, h5py.Dataset):
            base = Path(name).name.lower()
            if base in wanted or name.lower().replace("/", "_") in wanted:
                try:
                    return obj[()]
                except Exception:
                    return None
    return None


def _h5_find_numeric(h5, candidate_names):
    value = _h5_find_attr(h5, candidate_names)
    if value is None:
        value = _h5_find_dataset(h5, candidate_names)
    value = _as_python_scalar(value)
    try:
        return float(value)
    except Exception:
        return np.nan


def _parse_h5_filename(path, source_label):
    stem = Path(path).stem
    parts = stem.split("_")
    family = "xgb" if "xgb" in stem.lower() or "xgb" in str(source_label).lower() else "enet" if "enet" in stem.lower() or "enet" in str(source_label).lower() else str(source_label)
    protein = parts[0] if parts else stem

    normalization = None
    grid_preset = None
    select_k = np.nan

    for token in parts:
        low = token.lower()
        if low in {"none", "zscore", "ranking", "rank", "qbin50", "unknown"}:
            normalization = low
        if low in {"tiny", "fast", "current", "full"}:
            grid_preset = low
        if low.startswith("k") and low[1:].isdigit():
            select_k = int(low[1:])

    normalization = normalization or _h5_find_text_hint_from_name(stem, ["none", "zscore", "ranking", "rank", "qbin50", "unknown"])
    grid_preset = grid_preset or _h5_find_text_hint_from_name(stem, ["tiny", "fast", "current", "full"])

    run_bits = [str(source_label)]
    run_bits.append(normalization or "unknown")
    run_bits.append(grid_preset or "unknown")
    if pd.notna(select_k):
        run_bits.append(f"k{int(select_k)}")
    run_label = " | ".join(run_bits)

    return {
        "protein": protein,
        "model_family": family,
        "source_folder": str(source_label),
        "normalization": normalization,
        "grid_preset": grid_preset,
        "select_k": select_k,
        "run_label": run_label,
    }


def _h5_find_text_hint_from_name(stem, options):
    low = stem.lower()
    for opt in options:
        if f"_{opt}_" in f"_{low}_" or low.endswith(f"_{opt}"):
            return opt
    return None


def _safe_json_loads(value):
    if value is None:
        return None
    value = _as_python_scalar(value)
    if isinstance(value, (dict, list)):
        return value
    if not isinstance(value, str):
        return value
    try:
        return json.loads(value)
    except Exception:
        return value


def _fallback_read_test_samples_from_h5(path):
    candidate_names = [
        "test_samples", "test_sample_names", "test_index", "test_indices_names",
        "X_test_index", "y_test_index", "test_cell_lines", "loo_test_samples",
    ]
    with h5py.File(path, "r") as h5:
        value = _h5_find_attr(h5, candidate_names)
        if value is not None:
            return _decode_array(value)
        value = _h5_find_dataset(h5, candidate_names)
        if value is not None:
            return _decode_array(value)
    return None


def _fallback_load_h5_summary(path, source_label):
    path = Path(path)
    row = _parse_h5_filename(path, source_label)
    row.update({
        "h5_path": str(path),
        "file": path.name,
        "file_size_mb": path.stat().st_size / (1024 ** 2) if path.exists() else np.nan,
        "load_error": None,
    })
    try:
        with h5py.File(path, "r") as h5:
            # Prefer explicit attrs/datasets; fall back to NaN when absent.
            row["protein"] = _as_python_scalar(_h5_find_attr(h5, ["protein", "target", "target_protein"])) or row["protein"]
            row["test_mse"] = _h5_find_numeric(h5, ["test_mse", "mse_test", "mse"])
            row["test_pearson_corr"] = _h5_find_numeric(h5, ["test_pearson_corr", "test_pearson", "pearson_test", "pearson_corr", "pcc"])
            row["test_cosine_similarity"] = _h5_find_numeric(h5, ["test_cosine_similarity", "test_cosine", "cosine_similarity", "cosine"])
            row["test_r2"] = _h5_find_numeric(h5, ["test_r2", "r2_test", "r2", "test_r2_score"])
            row["selected_feature_count"] = _h5_find_numeric(h5, ["selected_feature_count", "n_selected_features", "selected_features"])
            row["train_n"] = _h5_find_numeric(h5, ["train_n", "n_train"])
            row["test_n"] = _h5_find_numeric(h5, ["test_n", "n_test"])
            row["best_params"] = _safe_json_loads(_h5_find_attr(h5, ["best_params", "params", "best_params_json"]))

            # If R2 is not stored but true/pred arrays are stored, compute it.
            if not np.isfinite(row.get("test_r2", np.nan)):
                y_true = _h5_find_dataset(h5, ["y_test", "test_true", "true", "y_true_test"])
                y_pred = _h5_find_dataset(h5, ["y_pred", "test_pred", "predicted", "y_pred_test"])
                if y_true is not None and y_pred is not None:
                    row["test_r2"] = _safe_r2_array(y_true, y_pred)
    except Exception as e:
        row["load_error"] = f"{type(e).__name__}: {e}"
    return row


def _fallback_add_family_group(df):
    out = df.copy()
    def family(row):
        text = " ".join(str(row.get(c, "")) for c in ["family_group", "model_family", "source_folder", "run_label"])
        text = text.lower()
        if "proxy" in text or "same_gene" in text:
            return "proxy"
        if "enet" in text or "elastic" in text:
            return "enet"
        if "xgb" in text or "xgboost" in text:
            return "xgb"
        return row.get("model_family", "unknown")
    out["family_group"] = out.apply(family, axis=1)
    return out


def _clean_pair(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float).reshape(-1)
    y_pred = np.asarray(y_pred, dtype=float).reshape(-1)
    mask = np.isfinite(y_true) & np.isfinite(y_pred)
    return y_true[mask], y_pred[mask]


def pearson_correlation(y_true, y_pred):
    y_true, y_pred = _clean_pair(y_true, y_pred)
    if len(y_true) <= 1 or np.std(y_true) <= 0 or np.std(y_pred) <= 0:
        return np.nan
    return float(np.corrcoef(y_true, y_pred)[0, 1])


def mse(y_true, y_pred):
    y_true, y_pred = _clean_pair(y_true, y_pred)
    if len(y_true) == 0:
        return np.nan
    return float(np.mean((y_true - y_pred) ** 2))


def rmse(y_true, y_pred):
    value = mse(y_true, y_pred)
    return float(np.sqrt(value)) if np.isfinite(value) else np.nan


def mae(y_true, y_pred):
    y_true, y_pred = _clean_pair(y_true, y_pred)
    if len(y_true) == 0:
        return np.nan
    return float(np.mean(np.abs(y_true - y_pred)))


def cosine_similarity(y_true, y_pred):
    y_true, y_pred = _clean_pair(y_true, y_pred)
    if len(y_true) == 0:
        return np.nan
    denom = np.linalg.norm(y_true) * np.linalg.norm(y_pred)
    if not np.isfinite(denom) or denom <= 0:
        return np.nan
    return float(np.dot(y_true, y_pred) / denom)


def _safe_r2_array(y_true, y_pred):
    y_true, y_pred = _clean_pair(y_true, y_pred)
    if len(y_true) <= 1:
        return np.nan
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    if ss_tot <= 0:
        return np.nan
    return float(1.0 - ss_res / ss_tot)


def _fallback_compute_metrics(y_true, y_pred):
    y_true_clean, y_pred_clean = _clean_pair(y_true, y_pred)
    return {
        "test_mse": mse(y_true_clean, y_pred_clean),
        "test_pearson_corr": pearson_correlation(y_true_clean, y_pred_clean),
        "test_cosine_similarity": cosine_similarity(y_true_clean, y_pred_clean),
        "test_r2": _safe_r2_array(y_true_clean, y_pred_clean),
        "n_test_used": len(y_true_clean),
    }


def _fallback_build_same_name_corr_lookup_fast(corr_obj, possible_same_name):
    # The same-gene correlation is metadata only for this notebook's proxy plots.
    possible_same_name = [str(p) for p in possible_same_name]
    lookup = {p: np.nan for p in possible_same_name}

    if isinstance(corr_obj, pd.DataFrame):
        df = corr_obj.copy()
        df.index = df.index.map(str)
        df.columns = df.columns.map(str)

        # Square matrix: use diagonal same-gene entries where available.
        if df.shape[0] == df.shape[1]:
            for p in possible_same_name:
                if p in df.index and p in df.columns:
                    try:
                        lookup[p] = float(df.loc[p, p])
                    except Exception:
                        pass
            return lookup

        # Long table: search common column names.
        low_cols = {c.lower(): c for c in df.columns}
        protein_col = low_cols.get("protein") or low_cols.get("target") or low_cols.get("target_protein")
        gene_col = low_cols.get("gene") or low_cols.get("feature") or low_cols.get("rna")
        corr_col = low_cols.get("pearson") or low_cols.get("correlation") or low_cols.get("corr") or low_cols.get("value")
        if protein_col and gene_col and corr_col:
            sub = df[df[protein_col].astype(str).eq(df[gene_col].astype(str))]
            for _, r in sub.iterrows():
                p = str(r[protein_col])
                if p in lookup:
                    try:
                        lookup[p] = float(r[corr_col])
                    except Exception:
                        pass
            return lookup

    if isinstance(corr_obj, dict):
        for p in possible_same_name:
            try:
                lookup[p] = float(corr_obj.get(p, np.nan))
            except Exception:
                pass

    return lookup


def _fallback_best_family(metric_col, family, higher=True, source_df_name="all_results"):
    df = globals()[source_df_name]
    if "family_group" not in df.columns:
        df = add_family_group(df)
    sub = df[df["family_group"].astype(str).eq(family)].copy()
    if metric_col not in sub.columns:
        return sub.iloc[0:0].copy()
    sub[metric_col] = pd.to_numeric(sub[metric_col], errors="coerce")
    sub = sub[np.isfinite(sub[metric_col])].copy()
    if sub.empty:
        return sub
    sub["protein"] = sub["protein"].astype(str)
    sub = sub.sort_values(["protein", metric_col], ascending=[True, not higher])
    return sub.drop_duplicates("protein", keep="first").reset_index(drop=True)


_define_if_missing("require_file", _fallback_require_file)
_define_if_missing("split_hash", _fallback_split_hash)
_define_if_missing("read_test_samples_from_h5", _fallback_read_test_samples_from_h5)
_define_if_missing("load_h5_summary", _fallback_load_h5_summary)
_define_if_missing("add_family_group", _fallback_add_family_group)
_define_if_missing("compute_metrics", _fallback_compute_metrics)
_define_if_missing("build_same_name_corr_lookup_fast", _fallback_build_same_name_corr_lookup_fast)
_define_if_missing("best_family", _fallback_best_family)
```

## Build all three same-gene proxy variants

For each same-name gene/protein pair, the notebook computes raw, train-z-score, and train-linear predictions in one protein loop.

```python
def zscore_gene_to_protein_scale(x_train, y_train, x_test):
    """Map test RNA z-scores to the training protein mean/SD scale."""
    x_train = np.asarray(x_train, dtype=float).reshape(-1)
    y_train = np.asarray(y_train, dtype=float).reshape(-1)
    x_test = np.asarray(x_test, dtype=float).reshape(-1)

    finite_x_train = np.isfinite(x_train)
    finite_y_train = np.isfinite(y_train)
    y_pred = np.full(x_test.shape, np.nan, dtype=float)

    if finite_y_train.sum() == 0 or finite_x_train.sum() == 0:
        return y_pred

    y_mean = float(np.mean(y_train[finite_y_train]))
    y_std = float(np.std(y_train[finite_y_train], ddof=0))
    x_mean = float(np.mean(x_train[finite_x_train]))
    x_std = float(np.std(x_train[finite_x_train], ddof=0))

    test_mask = np.isfinite(x_test)
    if not test_mask.any():
        return y_pred

    if not np.isfinite(x_std) or x_std <= 0:
        y_pred[test_mask] = y_mean
        return y_pred

    if not np.isfinite(y_std):
        y_std = 0.0

    gene_z = (x_test[test_mask] - x_mean) / x_std
    y_pred[test_mask] = y_mean + y_std * gene_z
    return y_pred


def linear_scale_prediction(x_train, y_train, x_test):
    """Fit protein ~ same-gene RNA on training samples and predict the test samples."""
    x_train = np.asarray(x_train, dtype=float).reshape(-1)
    y_train = np.asarray(y_train, dtype=float).reshape(-1)
    x_test = np.asarray(x_test, dtype=float).reshape(-1)

    train_mask = np.isfinite(x_train) & np.isfinite(y_train)
    y_pred = np.full(x_test.shape, np.nan, dtype=float)
    if train_mask.sum() == 0:
        return y_pred

    reg = LinearRegression()
    reg.fit(x_train[train_mask].reshape(-1, 1), y_train[train_mask])

    test_mask = np.isfinite(x_test)
    if test_mask.any():
        y_pred[test_mask] = reg.predict(x_test[test_mask].reshape(-1, 1))
    return y_pred


def predict_proxy_variant(variant, x_train, y_train, x_test):
    """Return predictions for one of the three merged proxy definitions."""
    if variant == "raw":
        return np.asarray(x_test, dtype=float).copy()
    if variant == "train_zscore":
        return zscore_gene_to_protein_scale(x_train, y_train, x_test)
    if variant == "train_linear":
        return linear_scale_prediction(x_train, y_train, x_test)
    raise ValueError(f"Unknown proxy variant: {variant}")


def proxy_best_params(variant):
    if variant == "raw":
        return {
            "baseline": "same_name_gene_raw_proxy",
            "normalization": "none",
            "calibration_model": "none",
            "test_split_source": "read_from_xgb_enet_h5",
        }
    if variant == "train_zscore":
        return {
            "baseline": "same_name_gene_train_zscore_proxy",
            "normalization": "gene_zscore_to_train_protein_scale",
            "calibration_fit_samples": "train_only",
            "test_split_source": "read_from_xgb_enet_h5",
        }
    if variant == "train_linear":
        return {
            "baseline": "same_name_gene_linear_regression_proxy",
            "calibration_model": "sklearn.linear_model.LinearRegression",
            "calibration_fit_samples": "train_only",
            "test_split_source": "read_from_xgb_enet_h5",
        }
    raise ValueError(f"Unknown proxy variant: {variant}")


def build_all_same_gene_proxy_results(variants=None):
    variants = list(variants or PROXY_VARIANT_ORDER)
    unknown = [v for v in variants if v not in PROXY_VARIANTS]
    if unknown:
        raise ValueError(f"Unknown proxy variants: {unknown}")

    rows = []
    possible = sorted(set(Y.columns.astype(str)).intersection(set(X.columns.astype(str))))
    pearson_lookup = build_same_name_corr_lookup_fast(pearson_corr_obj, possible)

    print("Same-name protein/gene pairs possible:", len(possible))
    print("Test samples used:", len(test_samples))
    print("Train samples used:", len(train_samples))
    print("Building variants:", variants)

    for protein in possible:
        y_true = Y.loc[test_samples, protein].to_numpy(dtype=float)
        x_train = X.loc[train_samples, protein].to_numpy(dtype=float)
        y_train = Y.loc[train_samples, protein].to_numpy(dtype=float)
        x_test = X.loc[test_samples, protein].to_numpy(dtype=float)
        corr_value = pearson_lookup.get(protein, np.nan)

        for variant in variants:
            spec = PROXY_VARIANTS[variant]
            y_pred = predict_proxy_variant(variant, x_train, y_train, x_test)
            metrics = compute_metrics(y_true, y_pred)

            rows.append({
                "protein": protein,
                "run_label": spec["run_label"],
                "model_family": "same_gene_proxy",
                "family_group": "proxy",
                "comparison_group": spec["comparison_group"],
                "proxy_variant": variant,
                "source_folder": "proxy_one_gene",
                "normalization": spec["normalization"],
                "h5_path": "",
                "proxy_type": "one_gene",
                "same_gene_corr_full_reference_only": float(corr_value) if np.isfinite(corr_value) else np.nan,
                "test_mse": metrics.get("test_mse", np.nan),
                "test_pearson_corr": metrics.get("test_pearson_corr", np.nan),
                "test_cosine_similarity": metrics.get("test_cosine_similarity", np.nan),
                "test_r2": metrics.get("test_r2", _safe_r2_array(y_true, y_pred)),
                "n_test_used": metrics.get("n_test_used", np.nan),
                "selected_feature_count": 1,
                "train_n": len(train_samples),
                "test_n": len(test_samples),
                "split_hash": model_split_hash,
                "best_params": proxy_best_params(variant),
            })

    return pd.DataFrame(rows)


proxy_results = build_all_same_gene_proxy_results()
proxy_results = add_family_group(proxy_results)

model_results = add_family_group(model_results)
model_results["comparison_group"] = model_results["family_group"].astype(str)
model_results["proxy_variant"] = np.nan

all_results = pd.concat([model_results, proxy_results], ignore_index=True, sort=False)

print("\nCombined result table")
print("- Saved model rows:", len(model_results))
print("- Proxy rows:      ", len(proxy_results))
print("- Total rows:      ", len(all_results))
print("- Unique proteins: ", all_results["protein"].nunique())

display(
    all_results.groupby(["comparison_group", "run_label"])["protein"]
    .nunique()
    .sort_values(ascending=False)
    .to_frame("proteins")
)

display(
    all_results.groupby("comparison_group")
    .agg(
        rows=("protein", "count"),
        proteins=("protein", "nunique"),
        pearson_nonmissing=("test_pearson_corr", lambda s: pd.to_numeric(s, errors="coerce").notna().sum()),
        mse_nonmissing=("test_mse", lambda s: pd.to_numeric(s, errors="coerce").notna().sum()),
        cosine_nonmissing=("test_cosine_similarity", lambda s: pd.to_numeric(s, errors="coerce").notna().sum()),
        r2_nonmissing=("test_r2", lambda s: pd.to_numeric(s, errors="coerce").notna().sum()),
    )
)

try:
    uc.set_module_globals(globals())
except Exception:
    pass
```

## Metric-wise best test performance tables

For ENet/XGBoost, the best saved run is selected separately for each metric and protein. Each proxy variant has one row per protein.

```python
def best_comparison_group(metric_col, group, higher=True):
    """Select the best saved row per protein for one comparison group and metric."""
    sub = all_results[all_results["comparison_group"].astype(str).eq(group)].copy()
    if metric_col not in sub.columns:
        return sub.iloc[0:0].copy()

    sub[metric_col] = pd.to_numeric(sub[metric_col], errors="coerce")
    sub = sub[np.isfinite(sub[metric_col])].copy()
    if sub.empty:
        return sub

    sub["protein"] = sub["protein"].astype(str)
    sub = sub.sort_values(["protein", metric_col], ascending=[True, not higher])
    return sub.drop_duplicates("protein", keep="first").reset_index(drop=True)


def best_group_metric_table(group):
    """Build a metric-wise best protein table for a proxy/model comparison group."""
    pieces = []
    selected_runs = []
    base_proteins = pd.Index([], dtype=str)

    metric_specs = [
        ("test_pearson_corr", "pearson", True),
        ("test_mse", "mse", False),
        ("test_cosine_similarity", "cosine", True),
        ("test_r2", "r2", True),
    ]

    for metric_col, out_col, higher in metric_specs:
        b = best_comparison_group(metric_col, group, higher=higher)
        if b.empty:
            continue
        keep = b[["protein", metric_col, "run_label"]].copy()
        keep["protein"] = keep["protein"].astype(str)
        keep = keep.rename(columns={metric_col: out_col, "run_label": f"{out_col}_run_label"})
        pieces.append(keep)
        base_proteins = base_proteins.union(pd.Index(keep["protein"].astype(str)))
        selected_runs.append((
            out_col,
            keep[f"{out_col}_run_label"].value_counts().rename_axis("run_label").reset_index(name="proteins"),
        ))

    if not pieces:
        return pd.DataFrame(columns=["protein", "pearson", "mse", "cosine", "r2"]), selected_runs

    out = pd.DataFrame({"protein": sorted(base_proteins.astype(str))})
    for keep in pieces:
        out = out.merge(keep, on="protein", how="left")

    for col in ["pearson", "mse", "cosine", "r2"]:
        if col not in out.columns:
            out[col] = np.nan
        out[col] = pd.to_numeric(out[col], errors="coerce")

    return out.sort_values("pearson", ascending=False, na_position="last").reset_index(drop=True), selected_runs


family_metric_tables = {}
family_selected_run_counts = {}
for group in [PROXY_VARIANTS[v]["comparison_group"] for v in PROXY_VARIANT_ORDER] + ["enet", "xgb"]:
    display_name = GROUP_TO_DISPLAY[group]
    table, run_counts = best_group_metric_table(group)
    family_metric_tables[display_name] = table
    family_selected_run_counts[display_name] = run_counts
    print(f"{display_name}: {len(table)} proteins with at least one valid metric")

print("\nMetric-specific selected run counts")
for display_name, counts in family_selected_run_counts.items():
    print(f"\n{display_name}")
    for metric_name, count_df in counts:
        print(f"  Best by {metric_name}:")
        display(count_df.head(10))
```

## Summary statistics and mean comparisons

```python
PERFORMANCE_METRICS = ["pearson", "mse", "cosine", "r2"]
COMPARISON_METRICS = ["pearson", "mse", "cosine"]
METRIC_DISPLAY_NAMES = {
    "pearson": "PCC",
    "mse": "MSE",
    "cosine": "Cosine Similarity",
    "r2": "R²",
}


def summarize_family_metrics(tables, model_order=None):
    model_order = list(model_order or tables.keys())
    rows = []
    for model, df in tables.items():
        for metric in PERFORMANCE_METRICS:
            if metric not in df.columns:
                continue
            values = pd.to_numeric(df[metric], errors="coerce").dropna()
            if len(values) == 0:
                continue
            rows.append({
                "model": model,
                "metric": metric,
                "metric_label": METRIC_DISPLAY_NAMES.get(metric, metric),
                "n": int(values.shape[0]),
                "mean": float(values.mean()),
                "std": float(values.std()),
                "median": float(values.median()),
                "min": float(values.min()),
                "max": float(values.max()),
            })
    out = pd.DataFrame(rows)
    if out.empty:
        return out
    out["model"] = pd.Categorical(out["model"], categories=model_order, ordered=True)
    out["metric"] = pd.Categorical(out["metric"], categories=PERFORMANCE_METRICS, ordered=True)
    return out.sort_values(["metric", "model"]).reset_index(drop=True)


def build_mean_comparisons(summary_df):
    rows = []
    higher_is_better = {"pearson": True, "mse": False, "cosine": True}
    for metric in COMPARISON_METRICS:
        sub = summary_df[(summary_df["metric"].astype(str) == metric) & np.isfinite(summary_df["mean"])].copy()
        if sub.empty:
            continue
        ranked = sub.sort_values("mean", ascending=not higher_is_better[metric]).reset_index(drop=True)
        best_model = ranked.loc[0, "model"]
        best_mean = ranked.loc[0, "mean"]
        for rank_idx, (_, row) in enumerate(ranked.iterrows(), start=1):
            rows.append({
                "metric": metric,
                "metric_label": METRIC_DISPLAY_NAMES.get(metric, metric),
                "higher_is_better": higher_is_better[metric],
                "model": row["model"],
                "mean": row["mean"],
                "rank": rank_idx,
                "best_model": best_model,
                "difference_vs_best": row["mean"] - best_mean,
            })
    return pd.DataFrame(rows)


def build_pairwise_mean_differences(summary_df, models=None):
    rows = []
    models = list(models or MODEL_DISPLAY_ORDER)
    for metric in COMPARISON_METRICS:
        means = summary_df[summary_df["metric"].astype(str).eq(metric)].set_index("model")["mean"].to_dict()
        for i, first in enumerate(models):
            for second in models[i + 1:]:
                first_value = means.get(first, np.nan)
                second_value = means.get(second, np.nan)
                rows.append({
                    "metric": metric,
                    "metric_label": METRIC_DISPLAY_NAMES.get(metric, metric),
                    "comparison": f"{first} - {second}",
                    "mean_difference": (
                        first_value - second_value
                        if np.isfinite(first_value) and np.isfinite(second_value)
                        else np.nan
                    ),
                })
    return pd.DataFrame(rows)


summary_stats = summarize_family_metrics(family_metric_tables, MODEL_DISPLAY_ORDER)
mean_comparisons = build_mean_comparisons(summary_stats)
pairwise_mean_differences = build_pairwise_mean_differences(summary_stats, MODEL_DISPLAY_ORDER)

print("Summary statistics by model/proxy group and test metric")
print("R² is retained wherever valid values are available.")
display(summary_stats)

print("Mean performance ranking by metric")
print("Cross-model rankings use PCC, MSE, and Cosine Similarity.")
display(mean_comparisons)

print("Pairwise differences of means")
print("For PCC/Cosine, positive favors the first model; for MSE, negative favors the first model.")
display(pairwise_mean_differences)
```

## Shared plotting functions

```python
def _add_panel_label(ax, label):
    ax.text(
        0.01, 0.99, label,
        transform=ax.transAxes,
        fontsize=18.8,
        fontweight="bold",
        va="top",
        ha="left",
        bbox=dict(boxstyle="round,pad=0.2", facecolor="white", edgecolor="black", alpha=0.9),
    )


def _get_group_colors(group="proteins"):
    # RAW ElasticNet palette from notebook 02.
    if group == "proteins":
        return {
            "pearson": "#F5B041",
            "mse": "#E74C3C",
            "cosine": "#5DADE2",
            "r2": "#AF7AC5",
        }
    elif group == "samples":
        return {
            "pearson": "#F8C471",
            "mse": "#EC7063",
            "cosine": "#85C1E9",
            "r2": "#C39BD3",
        }
    else:
        raise ValueError("group must be either 'samples' or 'proteins'")


def _style_axis(ax):
    ax.set_facecolor("#FFFFFF")
    ax.yaxis.grid(True, linestyle="--", alpha=0.3)
    ax.set_axisbelow(True)
    for spine in ax.spines.values():
        spine.set_color("#BDC3C7")
        spine.set_linewidth(1.2)
    ax.tick_params(axis="x", labelsize=14.5, colors="#2C3E50")
    ax.tick_params(axis="y", labelsize=14.5, colors="#7F8C8D")


def _plot_metric_hist(ax, values, title, xlabel, color, panel_label=None):
    values = pd.Series(values).dropna()
    ax.set_facecolor("#FFFFFF")
    if len(values) > 0:
        ax.hist(values, bins=30, color=color, edgecolor="black", alpha=0.8)
        mean_val = values.mean()
        if np.isfinite(mean_val):
            ax.axvline(mean_val, color="red", linestyle="--", linewidth=2)
    else:
        ax.text(0.5, 0.5, "No valid values", ha="center", va="center", transform=ax.transAxes)

    ax.set_xlabel(xlabel, fontsize=14.5)
    ax.set_ylabel("Count", fontsize=14.5)
    _remember_axes_name(ax, title)
    _style_axis(ax)
    if panel_label is not None:
        _add_panel_label(ax, panel_label)


def _format_metric_block(df, metric, label):
    if metric not in df.columns:
        return f"\n{label}:\n  No valid values\n"
    values = pd.to_numeric(df[metric], errors="coerce").dropna()
    if len(values) == 0:
        return f"\n{label}:\n  No valid values\n"
    return f"""
{label}:
  Mean: {values.mean():.4f} +/- {values.std():.4f}
  Min:  {values.min():.4f}
  Max:  {values.max():.4f}
"""


def _build_summary_text(df, title_label="Per-Protein", count_label="Proteins", metrics=None):
    if metrics is None:
        metrics = ["pearson", "mse", "cosine", "r2"]

    metric_labels = {
        "pearson": "Pearson",
        "mse": "MSE",
        "cosine": "Cosine Similarity",
        "r2": "R²",
    }

    text = f"""{title_label} Summary
{'='*28}
{count_label} (n={len(df)})
"""
    for metric in metrics:
        text += _format_metric_block(df, metric, metric_labels.get(metric, metric))
    return text


def _plot_best_worst_by_metric(
    ax,
    df,
    metric,
    xlabel,
    title,
    panel_label,
    higher_is_better=True,
    n_each=10,
):
    work = df[["protein", metric]].copy() if metric in df.columns else pd.DataFrame(columns=["protein", metric])
    work[metric] = pd.to_numeric(work[metric], errors="coerce")
    work = work.dropna(subset=[metric])

    if work.empty:
        ax.text(0.5, 0.5, "No valid values", ha="center", va="center", transform=ax.transAxes)
    else:
        best = work.sort_values(metric, ascending=not higher_is_better).head(n_each)
        worst = work.sort_values(metric, ascending=higher_is_better).head(n_each)

        # Sort each block so the horizontal bars read naturally.
        best = best.sort_values(metric, ascending=True)
        worst = worst.sort_values(metric, ascending=False)
        combined = pd.concat([best, worst], axis=0)
        bar_colors = (["#58D68D"] * len(best)) + (["#E74C3C"] * len(worst))
        y_pos = np.arange(len(combined))

        ax.barh(y_pos, combined[metric].values, color=bar_colors, edgecolor="black", alpha=0.8)
        ax.set_yticks(y_pos)
        ax.set_yticklabels(combined["protein"].astype(str).values, fontsize=11.6)

        if metric != "mse":
            ax.axvline(0, color="red", linestyle="-", linewidth=1)
        else:
            ax.axvline(work[metric].mean(), color="red", linestyle="--", linewidth=1.5)

    ax.set_xlabel(xlabel, fontsize=14.5)
    _remember_axes_name(ax, title)
    _style_axis(ax)
    _add_panel_label(ax, panel_label)


def plot_overall_multi_model(
    summary_df,
    out_path=None,
    title="Proxy vs ElasticNet vs XGBoost: Test-Set Performance",
    model_order=None,
):
    """1x3 overall mean bar plot for PCC, MSE, and Cosine."""
    model_order = list(model_order or MODEL_DISPLAY_ORDER)
    plt.style.use("seaborn-v0_8-whitegrid")
    colors = _get_group_colors("proteins")

    fig_width = max(15, 2.25 * len(model_order) + 5)
    fig, axes = plt.subplots(1, 3, figsize=readable_figsize((fig_width, 4.8)))
    fig.patch.set_facecolor("#F8F9FA")
    _remember_figure_name(fig, title)

    plot_specs = [
        ("pearson", "Pearson Correlation", "PCC", colors["pearson"], "a)"),
        ("mse", "Mean Squared Error", "MSE", colors["mse"], "b)"),
        ("cosine", "Cosine Similarity", "Cosine Similarity", colors["cosine"], "c)"),
    ]

    for ax, (metric, plot_title, ylabel, color, panel_label) in zip(axes.flatten(), plot_specs):
        sub = summary_df[summary_df["metric"].astype(str).eq(metric)].copy()
        sub["model"] = sub["model"].astype(str)
        sub = sub.set_index("model").reindex(model_order).reset_index()
        values = pd.to_numeric(sub["mean"], errors="coerce").to_numpy(dtype=float)
        labels = sub["model"].tolist()

        finite_values = values[np.isfinite(values)]
        if len(finite_values) > 0:
            ax.bar(
                labels,
                values,
                color=color,
                width=0.6,
                edgecolor="#2C3E50",
                linewidth=2,
                alpha=0.9,
            )
            if metric == "mse":
                ymin = 0 if np.nanmin(values) >= 0 else np.nanmin(values) * 1.15
                ymax = np.nanmax(values) * 1.25 if np.nanmax(values) != 0 else 1
                ax.set_ylim(ymin, ymax)
            else:
                ymin = min(0, np.nanmin(values) * 1.2)
                ymax = np.nanmax(values) * 1.25 if np.nanmax(values) > 0 else 1
                ax.set_ylim(ymin, ymax)
        else:
            ax.text(0.5, 0.5, "No valid values", ha="center", va="center", transform=ax.transAxes)

        _remember_axes_name(ax, plot_title)
        ax.set_ylabel(ylabel, fontsize=15.9, fontweight="bold", color="#2C3E50")
        ax.tick_params(axis="x", rotation=25 if len(model_order) > 3 else 15)
        _style_axis(ax)
        _add_panel_label(ax, panel_label)

    plt.tight_layout(rect=[0, 0, 1, 0.94])

    if out_path is not None:
        save_readable_current_figure(out_path, dpi=200, bbox_inches="tight", facecolor=fig.get_facecolor())
        print(f"Saved: {out_path}")

    show_readable_figures()


def plot_detailed_model_proxy(df, model_label="Proxy", out_path=None):
    """2x3 RAW-style detailed test-set performance plot for Proxy, keeping R²."""
    plt.style.use("seaborn-v0_8-whitegrid")
    colors = _get_group_colors("proteins")
    df = df.copy()

    fig = plt.figure(figsize=readable_figsize((22, 11)))
    fig.patch.set_facecolor("#F8F9FA")
    gs = fig.add_gridspec(2, 3, height_ratios=[1, 1.2])
    _remember_figure_name(fig, f"{model_label}: Test-Set Protein-Wise Performance")

    ax1 = fig.add_subplot(gs[0, 0])
    _plot_metric_hist(ax1, df["pearson"], "Per-Protein PCC", "Pearson Correlation", colors["pearson"], panel_label="a)")

    ax2 = fig.add_subplot(gs[0, 1])
    _plot_metric_hist(ax2, df["mse"], "Per-Protein MSE", "MSE", colors["mse"], panel_label="b)")

    ax3 = fig.add_subplot(gs[0, 2])
    _plot_metric_hist(ax3, df["cosine"], "Per-Protein Cosine Similarity", "Cosine Similarity", colors["cosine"], panel_label="c)")

    ax4 = fig.add_subplot(gs[1, 0])
    _plot_metric_hist(ax4, df["r2"], "Per-Protein R²", "R² Score", colors["r2"], panel_label="d)")

    ax5 = fig.add_subplot(gs[1, 1])
    _plot_best_worst_by_metric(
        ax5,
        df,
        metric="pearson",
        xlabel="PCC",
        title="Best 10 / Worst 10 Proteins by PCC",
        panel_label="e)",
        higher_is_better=True,
    )

    ax6 = fig.add_subplot(gs[1, 2])
    ax6.axis("off")
    ax6.set_facecolor("#FFFFFF")
    _remember_axes_name(ax6, "Summary Statistics")
    _add_panel_label(ax6, "f)")

    summary_text = _build_summary_text(df=df, title_label="Per-Protein", count_label="Proteins", metrics=["pearson", "mse", "cosine", "r2"])
    ax6.text(
        0.5, 0.92, summary_text,
        transform=ax6.transAxes,
        fontsize=13.0,
        va="top",
        ha="center",
        multialignment="left",
        fontfamily="monospace",
        bbox=dict(boxstyle="round", facecolor="wheat", alpha=0.45),
    )

    plt.tight_layout(rect=[0, 0, 1, 0.96])

    if out_path is not None:
        save_readable_current_figure(out_path, dpi=150, bbox_inches="tight", facecolor=fig.get_facecolor())
        print(f"Saved: {out_path}")

    show_readable_figures()


def plot_detailed_model_no_r2(df, model_label, out_path=None):
    """2x3 RAW-style detailed plot for ENet/XGB without R², adding best/worst MSE proteins."""
    plt.style.use("seaborn-v0_8-whitegrid")
    colors = _get_group_colors("proteins")
    df = df.copy()

    fig = plt.figure(figsize=readable_figsize((22, 11)))
    fig.patch.set_facecolor("#F8F9FA")
    gs = fig.add_gridspec(2, 3, height_ratios=[1, 1.2])
    _remember_figure_name(fig, f"{model_label}: Test-Set Protein-Wise Performance")

    ax1 = fig.add_subplot(gs[0, 0])
    _plot_metric_hist(ax1, df["pearson"], "Per-Protein PCC", "Pearson Correlation", colors["pearson"], panel_label="a)")

    ax2 = fig.add_subplot(gs[0, 1])
    _plot_metric_hist(ax2, df["mse"], "Per-Protein MSE", "MSE", colors["mse"], panel_label="b)")

    ax3 = fig.add_subplot(gs[0, 2])
    _plot_metric_hist(ax3, df["cosine"], "Per-Protein Cosine Similarity", "Cosine Similarity", colors["cosine"], panel_label="c)")

    ax4 = fig.add_subplot(gs[1, 0])
    _plot_best_worst_by_metric(
        ax4,
        df,
        metric="pearson",
        xlabel="PCC",
        title="Best 10 / Worst 10 Proteins by PCC",
        panel_label="d)",
        higher_is_better=True,
    )

    ax5 = fig.add_subplot(gs[1, 1])
    _plot_best_worst_by_metric(
        ax5,
        df,
        metric="mse",
        xlabel="MSE",
        title="Best 10 / Worst 10 Proteins by MSE",
        panel_label="e)",
        higher_is_better=False,
    )

    ax6 = fig.add_subplot(gs[1, 2])
    ax6.axis("off")
    ax6.set_facecolor("#FFFFFF")
    _remember_axes_name(ax6, "Summary Statistics")
    _add_panel_label(ax6, "f)")

    summary_text = _build_summary_text(df=df, title_label="Per-Protein", count_label="Proteins", metrics=["pearson", "mse", "cosine"])
    ax6.text(
        0.5, 0.92, summary_text,
        transform=ax6.transAxes,
        fontsize=13.0,
        va="top",
        ha="center",
        multialignment="left",
        fontfamily="monospace",
        bbox=dict(boxstyle="round", facecolor="wheat", alpha=0.45),
    )

    plt.tight_layout(rect=[0, 0, 1, 0.96])

    if out_path is not None:
        save_readable_current_figure(out_path, dpi=150, bbox_inches="tight", facecolor=fig.get_facecolor())
        print(f"Saved: {out_path}")

    show_readable_figures()
```

## Compact protein-metric figures and tables

These extra figures come after all earlier plots. Each has one row of three protein-wise distributions: **PCC, MSE and cosine similarity**, with no titles. Red dashed lines mark metric means. Best/worst proteins and summary statistics are printed as tables instead of occupying plot panels. CSV copies are saved in `figures_pdf/tables`.

The final export cell saves the compact PDFs using the stable filenames shown below, without a numerical prefix. The original larger plots remain available as title-free versions. No models are refitted and no metrics are recomputed.


```python
# Compact protein-wise distributions and separate data tables.
# These functions reuse the metric tables computed earlier in the notebook.
COMPACT_TABLE_DIR = FIGURES_PDF_DIR / 'tables'
COMPACT_TABLES = {}


def print_compact_protein_tables(protein_metrics, model_label, file_stem):
    """Print and save summary statistics and metric-specific best/worst proteins."""
    data = protein_metrics.copy()
    if 'protein' not in data.columns:
        if data.index.name == 'protein':
            data = data.reset_index()
        else:
            raise ValueError(f'{model_label}: expected a protein column.')
    metric_specs = [
        ('pearson', 'PCC', True),
        ('mse', 'MSE', False),
        ('cosine', 'Cosine similarity', True),
    ]
    required = ['protein'] + [metric for metric, _, _ in metric_specs]
    missing = [column for column in required if column not in data.columns]
    if missing:
        raise ValueError(f'{model_label}: missing columns {missing}')

    summary_rows = []
    rankings = {}
    for metric, display_name, higher_is_better in metric_specs:
        values = pd.to_numeric(data[metric], errors='coerce').replace([np.inf, -np.inf], np.nan)
        data[metric] = values
        valid = values.dropna()
        summary_rows.append({
            'metric': display_name,
            'n_valid': len(valid),
            'n_missing_or_nonfinite': len(values) - len(valid),
            'mean': valid.mean(),
            'std': valid.std(),
            'min': valid.min(),
            'q25': valid.quantile(0.25),
            'median': valid.median(),
            'q75': valid.quantile(0.75),
            'max': valid.max(),
        })
        ordered = data.loc[values.notna(), ['protein', metric]].sort_values(
            metric, ascending=not higher_is_better, kind='stable',
        )
        best = ordered.head(10).copy()
        worst = ordered.sort_values(metric, ascending=higher_is_better, kind='stable').head(10).copy()
        best.insert(0, 'set', 'Best')
        worst.insert(0, 'set', 'Worst')
        best.insert(1, 'rank', np.arange(1, len(best) + 1))
        worst.insert(1, 'rank', np.arange(1, len(worst) + 1))
        rankings[metric] = pd.concat([best, worst], ignore_index=True)

    summary = pd.DataFrame(summary_rows)
    COMPACT_TABLE_DIR.mkdir(parents=True, exist_ok=True)
    summary.to_csv(COMPACT_TABLE_DIR / f'{file_stem}_summary_statistics.csv', index=False)
    print('\n' + '=' * 80)
    print(model_label)
    print('Summary statistics:')
    print(summary.to_string(index=False, float_format=lambda value: f'{value:.6g}'))
    for metric, display_name, _ in metric_specs:
        table = rankings[metric]
        print(f'\nBest / worst proteins by {display_name}:')
        print(table.to_string(index=False, float_format=lambda value: f'{value:.6g}'))
        table.to_csv(COMPACT_TABLE_DIR / f'{file_stem}_best_worst_{metric}.csv', index=False)
    result = {'summary': summary, 'rankings': rankings}
    COMPACT_TABLES[file_stem] = result
    return result


def plot_compact_protein_metrics(protein_metrics, model_label, file_stem):
    """One title-free row: protein PCC, MSE and cosine-similarity distributions."""
    print_compact_protein_tables(protein_metrics, model_label, file_stem)
    colors = _get_group_colors('proteins')
    fig, axes = plt.subplots(1, 3, figsize=(15.5, 2.6))
    fig.patch.set_facecolor('white')
    fig._compact_protein_metrics = True
    fig._readable_pdf_pad = 0.04
    fig._readable_export_name = f'{file_stem}.pdf'
    for column, (metric, xlabel) in enumerate([
        ('pearson', 'Protein PCC'),
        ('mse', 'Protein MSE'),
        ('cosine', 'Protein cosine similarity'),
    ]):
        ax = axes[column]
        values = pd.to_numeric(protein_metrics[metric], errors='coerce')
        values = values[np.isfinite(values)]
        if len(values):
            ax.hist(values, bins=30, color=colors[metric], edgecolor='black', linewidth=0.6, alpha=0.85)
            ax.axvline(values.mean(), color='red', linestyle='--', linewidth=1.6)
        else:
            ax.text(0.5, 0.5, 'No finite values', transform=ax.transAxes,
                    ha='center', va='center', fontsize=16)
        ax.set_xlabel(xlabel, fontsize=18, labelpad=5)
        ax.set_ylabel('Proteins' if column == 0 else '', fontsize=17, labelpad=5)
        ax.tick_params(axis='both', labelsize=15.5, pad=3)
        ax.xaxis.set_major_locator(MaxNLocator(nbins=4))
        ax.yaxis.set_major_locator(MaxNLocator(nbins=3, integer=True))
        ax.grid(axis='y', linestyle='--', alpha=0.20)
        ax.set_axisbelow(True)
        ax.set_title('')
    prepare_readable_figure(fig)
    _cache_figure_pdf(fig, source_name=file_stem)
    show_readable_figures()
    # The in-memory PDF is retained for the last export cell.
    plt.close(fig)
    return fig
```

```python
compact_models = [
    ('Raw Proxy', 'figS2a_proxy_raw_metrics_1x3'),
    ('Z-score Proxy', 'figS2b_proxy_zscore_metrics_1x3'),
    ('Linear Proxy', 'figS2c_proxy_linear_metrics_1x3'),
    ('ElasticNet', 'figS2d_elasticnet_test_metrics_1x3'),
    ('XGBoost', 'figS2e_xgboost_test_metrics_1x3'),
]
for model_label, file_stem in compact_models:
    plot_compact_protein_metrics(
        family_metric_tables[model_label],
        model_label=model_label,
        file_stem=file_stem,
    )
print('Compact table CSV files:', COMPACT_TABLE_DIR)
```
