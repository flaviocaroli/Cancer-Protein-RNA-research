```python
import numpy as np
import pandas as pd
from joblib import load
import json
import os
```

## Converting to h5 and exporting results

```python
h5_path = 'data/elasticnet_results.h5'

with pd.HDFStore(h5_path, mode='w', complevel=7, complib='blosc') as store:
    for file in os.listdir('data'):
        if file.endswith('_fixed_features_elasticnet_LOO_results.joblib'):
            name = file.split('_')[0]
            df = load('data/' + file)
            df["params_dict"] = df["params_dict"].apply(json.dumps)
            store.put(key=name, value=df, format='table')
```
