# Binning, Ranking, and Z-score transforms for omics matrices

This notebook takes a **wide matrix** like the one in your screenshot:

- **rows = samples** (e.g., `MDAMB468_BREAST`)
- **columns = proteins / genes** (e.g., `A2M`, `AAAS`, ...)
- **values = numeric measurements**

It then computes three useful representations:

## 1) **Z-score normalization** (feature-wise / column-wise)
For each feature:
- subtract the feature mean
- divide by the feature standard deviation

## 2) Quantile **binning** (with a special bin for zeros)
For each sample:
- values equal to **0** → bin **0**
- all **nonzero** values → split into **b quantiles** → bins **1..b**

## 3) **Ranking**
For each sample:
- sort features by value and assign **ranks** (ties get the average rank)

These transforms are useful when you want models to depend more on **within-sample relative patterns**
than on absolute scales (which can drift across batches/datasets).

```python
# If you run this in a fresh environment, you may need:
# !pip install pandas numpy matplotlib pyarrow

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from joblib import load

np.random.seed(0)
```

## Plot styling and helper functions

These helpers make the charts easier to read and a bit more presentation-ready.


```python
plt.rcParams.update({
    "figure.figsize": (9, 5.2),
    "figure.dpi": 120,
    "axes.facecolor": "#fbfbfe",
    "figure.facecolor": "white",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "grid.alpha": 0.22,
    "grid.linestyle": "--",
    "axes.titleweight": "bold",
    "axes.titlesize": 14,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
})

COLORS = {
    "raw": "#5B5FDE",
    "zscore": "#00A884",
    "binned": "#F39C12",
    "ranked": "#D63384",
    "accent": "#17A2B8",
    "dark": "#2D3436",
    "muted": "#7F8C8D",
}


def sample_flat_values(df: pd.DataFrame, max_points: int = 200_000) -> np.ndarray:
    x = df.to_numpy().ravel()
    x = x[~np.isnan(x)]
    if x.size > max_points:
        idx = np.random.choice(x.size, size=max_points, replace=False)
        x = x[idx]
    return x


def _annotate_stats(ax, values, color_mean="#E74C3C", color_median="#16A085"):
    values = np.asarray(values, dtype=float)
    values = values[~np.isnan(values)]
    if values.size == 0:
        return
    mean_v = float(np.mean(values))
    median_v = float(np.median(values))
    ax.axvline(mean_v, color=color_mean, linestyle="--", linewidth=1.8, alpha=0.95, label=f"mean = {mean_v:.2f}")
    ax.axvline(median_v, color=color_median, linestyle=":", linewidth=2.2, alpha=0.95, label=f"median = {median_v:.2f}")
    ax.legend(frameon=False, loc="best")


def plot_distribution(values: np.ndarray, title: str, xlabel: str = "Value", bins: int = 60, color: str = "#5B5FDE"):
    values = np.asarray(values, dtype=float)
    values = values[~np.isnan(values)]

    fig, ax = plt.subplots()
    ax.hist(values, bins=bins, color=color, alpha=0.82, edgecolor="white", linewidth=0.7)
    #_annotate_stats(ax, values)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Count")
    plt.tight_layout()
    plt.show()


def plot_discrete_distribution(values: np.ndarray, title: str, xlabel: str, color: str):
    values = np.asarray(values, dtype=float)
    values = values[~np.isnan(values)]
    uniq, counts = np.unique(values, return_counts=True)

    fig, ax = plt.subplots()
    ax.bar(uniq, counts, width=0.85, color=color, alpha=0.88, edgecolor="white", linewidth=0.8)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Count")
    if len(uniq) <= 25:
        ax.set_xticks(uniq)
    plt.tight_layout()
    plt.show()


def plot_overlay_hist(values_a: np.ndarray, values_b: np.ndarray, label_a: str, label_b: str, title: str, xlabel: str,
                      bins: int = 80, color_a: str = "#5B5FDE", color_b: str = "#00A884"):
    values_a = np.asarray(values_a, dtype=float)
    values_b = np.asarray(values_b, dtype=float)
    values_a = values_a[~np.isnan(values_a)]
    values_b = values_b[~np.isnan(values_b)]

    fig, ax = plt.subplots()
    ax.hist(values_a, bins=bins, density=True, alpha=0.48, color=color_a, label=label_a, edgecolor="white", linewidth=0.5)
    ax.hist(values_b, bins=bins, density=True, alpha=0.48, color=color_b, label=label_b, edgecolor="white", linewidth=0.5)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Density")
    ax.legend(frameon=False)
    plt.tight_layout()
    plt.show()


def plot_scatter(x: np.ndarray, y: np.ndarray, title: str, xlabel: str, ylabel: str, color: str = "#17A2B8"):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    mask = (~np.isnan(x)) & (~np.isnan(y))
    x = x[mask]
    y = y[mask]

    fig, ax = plt.subplots()
    ax.scatter(x, y, s=14, alpha=0.45, color=color, edgecolors="none")
    if x.size > 1:
        order = np.argsort(x)
        x_sorted = x[order]
        y_sorted = y[order]
        window = max(15, min(301, (len(x_sorted) // 25) * 2 + 1))
        smooth = pd.Series(y_sorted).rolling(window=window, center=True, min_periods=max(5, window // 5)).median()
        ax.plot(x_sorted, smooth, color=COLORS["dark"], linewidth=2.1, alpha=0.9)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    plt.tight_layout()
    plt.show()


def plot_summary_boxplot(series_dict: dict, title: str, ylabel: str):
    labels = list(series_dict.keys())
    data = [pd.Series(v).dropna().to_numpy() for v in series_dict.values()]
    palette = [COLORS["ranked"], COLORS["binned"], COLORS["zscore"]][:len(data)]

    fig, ax = plt.subplots(figsize=(8, 5.2))
    box = ax.boxplot(data, labels=labels, patch_artist=True, widths=0.55)
    for patch, color in zip(box["boxes"], palette):
        patch.set_facecolor(color)
        patch.set_alpha(0.65)
        patch.set_edgecolor("#2D3436")
    for median in box["medians"]:
        median.set_color("#2D3436")
        median.set_linewidth(2)
    ax.set_title(title)
    ax.set_ylabel(ylabel)
    plt.tight_layout()
    plt.show()
```

## Load data

Supported formats: **.csv**, **.tsv/.txt**, **.parquet**, **.pkl**, **.joblib**

### Expected shape
- samples × features (e.g., 369 × 5113)

If the first column is sample IDs (strings), we’ll automatically use it as the index.


```python
DATA_PATH = "data/transcriptomics_filtered.joblib"  # <-- set this to your file path, e.g. "data/proteomics_matrix.csv"

def load_matrix(path: str) -> pd.DataFrame:
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"File not found: {path.resolve()}")

    ext = path.suffix.lower()

    if ext == ".csv":
        df = pd.read_csv(path)
    elif ext == ".joblib":
        df = load(path)
    elif ext in [".tsv", ".txt"]:
        df = pd.read_csv(path, sep="	")
    elif ext == ".parquet":
        df = pd.read_parquet(path)
    elif ext in [".pkl", ".pickle"]:
        df = pd.read_pickle(path)
    else:
        raise ValueError(f"Unsupported extension: {ext}")

    if "samples" in df.columns:
        df = df.set_index("samples")
    else:
        first = df.columns[0]
        if str(first).startswith("Unnamed") or str(first).strip() == "":
            df = df.set_index(first)
        else:
            if not pd.api.types.is_numeric_dtype(df[first]):
                numeric_ratio = df.drop(columns=[first]).apply(pd.to_numeric, errors="coerce").notna().mean().mean()
                if numeric_ratio > 0.95:
                    df = df.set_index(first)

    df = df.apply(pd.to_numeric, errors="coerce")
    return df

# Load
df = load_matrix(DATA_PATH)
df.shape, df.index.name, df.columns.name
```

## Optional preprocessing / additional view: z-score normalization

**Put this cell right here:** after the sanity checks and before the binning / ranking section.

This is the cleanest place because:
- the raw data has already been loaded and checked
- the z-scored matrix can now be plotted, saved, or fed into downstream steps
- it keeps preprocessing grouped together near the top of the notebook

This is **column-wise z-score normalization** (feature-wise across samples), which is the most common setup for ML tables.


```python
def zscore_by_feature(df: pd.DataFrame) -> pd.DataFrame:
    """Column-wise z-score normalization (feature-wise across samples)."""
    mu = df.mean(axis=0, skipna=True)
    sigma = df.std(axis=0, ddof=0, skipna=True).replace(0, np.nan)
    return (df - mu) / sigma


df_z = zscore_by_feature(df)

flat_z = sample_flat_values(df_z)
print(f"Z-scored matrix mean/std (global): {np.nanmean(flat_z):.4f} / {np.nanstd(flat_z):.4f}")
df_z.head()
```

## Core transforms: binning and ranking

### Binning details (matches the “quantile bins per sample” idea)
- We treat exact zeros as a special state: **bin 0**
- For nonzero values, we assign **approximately equal-frequency bins** using within-sample ranks.

This rank-based bin assignment is robust even when many values are identical (ties).


```python
def bin_by_quantiles_per_sample(
    df: pd.DataFrame,
    b: int = 20,
    zero_bin: bool = True,
) -> pd.DataFrame:
    """Row-wise quantile binning."""
    arr = df.to_numpy(dtype=float)
    out = np.full(arr.shape, np.nan, dtype=float)

    for i in range(arr.shape[0]):
        x = arr[i, :]
        nanmask = np.isnan(x)

        if zero_bin:
            zeromask = (x == 0) & (~nanmask)
        else:
            zeromask = np.zeros_like(x, dtype=bool)

        nzmask = (~nanmask) & (~zeromask)
        nz = x[nzmask]

        out[i, zeromask] = 0.0

        if nz.size == 0:
            continue

        ranks = pd.Series(nz).rank(method="average", ascending=True).to_numpy()
        pct = (ranks - 0.5) / nz.size
        bins = (np.floor(pct * b).astype(int) + 1).clip(1, b)
        out[i, nzmask] = bins.astype(float)

    return pd.DataFrame(out, index=df.index, columns=df.columns)


def rank_per_sample(df: pd.DataFrame) -> pd.DataFrame:
    """Row-wise ranks (ascending). Ties get average rank. NaNs preserved."""
    return df.rank(axis=1, method="average", ascending=True, na_option="keep")


B = 20  # <-- tune this (e.g., 5, 10, 20, 50, 100)
binned = bin_by_quantiles_per_sample(df, b=B, zero_bin=True)
ranked = rank_per_sample(df)

binned.shape, ranked.shape
```

## How much information is retained? (simple diagnostics)

- **Spearman correlation** between original values and ranks within each sample should be ~1
  (because rank is a monotonic transform).
- **Binning** is coarser, so the correlation is still positive but usually lower.

We compute a per-sample Spearman correlation using pandas.


```python
def spearman_per_row(a: pd.DataFrame, b: pd.DataFrame) -> pd.Series:
    out = []
    for idx in a.index:
        x = a.loc[idx]
        y = b.loc[idx]
        mask = x.notna() & y.notna()
        if mask.sum() < 3:
            out.append(np.nan)
        else:
            out.append(x[mask].corr(y[mask], method="spearman"))
    return pd.Series(out, index=a.index)


rho_rank = spearman_per_row(df, ranked)
rho_bin = spearman_per_row(df, binned)

print("Per-sample Spearman (original vs ranked):")
print(rho_rank.describe())

print("Per-sample Spearman (original vs binned):")
print(rho_bin.describe())

plot_distribution(rho_rank.dropna().to_numpy(), "Per-sample Spearman: original vs ranked", xlabel="Spearman rho", bins=40, color=COLORS["ranked"])
plot_distribution(rho_bin.dropna().to_numpy(), "Per-sample Spearman: original vs binned", xlabel="Spearman rho", bins=40, color=COLORS["binned"])
plot_summary_boxplot(
    {
        "Original vs ranked": rho_rank,
        "Original vs binned": rho_bin,
    },
    title="Per-sample Spearman comparison",
    ylabel="Spearman rho",
)
```

## Demonstration of invariance to monotonic scaling (optional but illuminating)

If we apply a **monotonic transform** to a sample like:

\[ x' = 2x + 1 \]

then the **rank ordering** is unchanged (so ranks stay identical), while the raw distribution shifts.
This is a toy illustration of why ranking can be robust to certain dataset-specific scaling effects.


```python
x2 = 2 * x + 1
mask2 = ~np.isnan(x2)

plot_overlay_hist(
    x[mask_x],
    x2[mask2],
    label_a="Original",
    label_b="x' = 2x + 1",
    title=f"Sample {SAMPLE_ID}: monotonic transform changes scale, not order",
    xlabel="Value",
    bins=80,
    color_a=COLORS["raw"],
    color_b=COLORS["accent"],
)

r1 = pd.Series(x).rank(method="average", ascending=True)
r2 = pd.Series(x2).rank(method="average", ascending=True)
same = (r1[mask_x] == r2[mask_x]).mean()

print(f"Fraction of features with identical rank after monotonic transform: {same:.4f}")
```

## Notes / tips

- If your matrix is already z-scored, **ranking/binning** still works.
- If you have many zeros, bin 0 can be meaningful (undetected vs low).
- **Choosing B**:
  - small B (e.g., 5–20): stronger robustness, less detail
  - large B (e.g., 50–200): closer to full ranking, more detail
- If you’re going to train models, compute any feature standardization using **training data only**.
- If you want to apply binning / ranking to the z-scored matrix instead of the raw matrix, replace `df` with `df_z` in the transform cell.
