```python
import os
import pandas as pd
import numpy as np
from sklearn.linear_model import ElasticNet
from single_pipeline import single_protein_pipeline

# Define paths
current_dir = os.getcwd()
data_dir = os.path.join(current_dir, "..", "data")
transcriptomics_file = os.path.join(data_dir, "transcriptomics_filtered.csv")
proteomics_file = os.path.join(data_dir, "proteomics_matched.csv")
protein_list_file = os.path.join(data_dir, "protein_list.txt")

# Load data
print("Loading data...")
try:
    transcriptomics = pd.read_csv(transcriptomics_file, index_col=0)  # samples x features
    proteomics = pd.read_csv(proteomics_file, index_col=0)            # samples x proteins
    with open(protein_list_file, 'r') as f:
        protein_name = f.readline().strip()  # Get the first protein from the file
except Exception as e:
    print(f"Error loading data: {e}")
    raise

# Check if protein exists in proteomics data
if protein_name not in proteomics.columns:
    raise ValueError(f"Protein {protein_name} not found in proteomics data.")

print(f"Selected protein: {protein_name}")

# Define the ElasticNet model and parameter grid
elasticnet_model = ElasticNet()
param_grid_elasticnet = {
    'alpha': [0.1, 0.3],          # Regularization strengths
    'l1_ratio': [0.1, 0.5, 0.9],  # Mix of L1 and L2
    'p_value': [0.1, 0.05, 0.01]  # P-values for feature selection
}

print(f"Running ElasticNet pipeline for protein {protein_name} over all samples (LOO)...")
try:
    # Run LOO pipeline for the single protein
    results_df = single_protein_pipeline(
        model=elasticnet_model,
        param_grid=param_grid_elasticnet,
        X=transcriptomics,
        y=proteomics,
        protein_name=protein_name,
        feature_selector='anova',
        n_splits=5
    )
except Exception as e:
    print(f"Error during pipeline execution: {e}")
    raise

# Save results
output_file = os.path.join(data_dir, f"{protein_name}_LOO_results.csv")
try:
    print(f"Saving results to {output_file}...")
    results_df.to_csv(output_file, index=False)
    print("Pipeline finished successfully.")
except Exception as e:
    print(f"Error saving results: {e}")
    raise
```
