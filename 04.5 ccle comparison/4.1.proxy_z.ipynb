# 4.5. Procedural CCLE proxy z-normalized protein prediction partitions

This notebook keeps `utils_ccle.py` unchanged. Any extra helper functions are defined directly inside this notebook.

The same-gene proxy for each protein is now **not a fitted regression model**. It uses train-only distribution matching:

```python
gene_z = (gene_test - mean(gene_train)) / std(gene_train)
predicted_protein = mean(protein_train) + std(protein_train) * gene_z
```

Only training samples are used to compute the gene and protein means/standard deviations, so the test set is not leaked into the proxy normalization.

**Added scTranslator rescue extension.** The final-rescue theme now covers not only Pearson correlation, but also cosine similarity and MSE. The added section asks whether scTranslator rescues proteins that remain unresolved after the base proxy -> ENet -> XGB cascade under each metric-specific threshold.

## Notebook-local proxy z-normalization helpers

The utility module is imported only for shared loading, plotting, and cascade helpers. The proxy z-normalization functions below intentionally override the imported proxy helpers inside the notebook namespace, without editing `utils_ccle.py`.


````python
# Notebook-local proxy z-normalization functions.
# These override the imported names only inside this notebook session.
# The utils_ccle.py file is not edited.


def _notebook_index_by_protein(df):
    """Small local version used by notebook-local cascade helpers."""
    if df is None or len(df) == 0:
        return pd.DataFrame().rename_axis("protein")
    out = df.copy()
    out["protein"] = out["protein"].astype(str)
    return out.drop_duplicates("protein", keep="first").set_index("protein")


def _notebook_proxy_label(row):
    """Map proxy row metadata to the two displayed proxy stages."""
    proxy_type = str(row.get("proxy_type", "")).lower()
    run_label = str(row.get("run_label", "")).lower()
    if "spearman" in proxy_type or "spearman" in run_label:
        return "Proxy Spearman"
    return "Proxy Pearson"


def zscore_gene_to_protein_scale(x_train, y_train, x_test):
    """
    Train-only same-gene proxy z-normalization.

    For each same-name gene/protein pair:
        gene_z = (gene_test - mean(gene_train)) / std(gene_train)
        predicted_protein = mean(protein_train) + std(protein_train) * gene_z

    This is distribution matching, not LinearRegression. No fitted regression
    slope/intercept is learned. Non-finite train values are ignored when
    computing train means/stds. Non-finite test RNA values remain NaN.
    """
    x_train = np.asarray(x_train, dtype=float).reshape(-1)
    y_train = np.asarray(y_train, dtype=float).reshape(-1)
    x_test = np.asarray(x_test, dtype=float).reshape(-1)

    finite_x_train = np.isfinite(x_train)
    finite_y_train = np.isfinite(y_train)

    y_pred = np.full(x_test.shape, np.nan, dtype=float)

    if finite_y_train.sum() == 0:
        return y_pred

    y_mean = float(np.mean(y_train[finite_y_train]))
    y_std = float(np.std(y_train[finite_y_train], ddof=0))

    if finite_x_train.sum() == 0:
        return y_pred

    x_mean = float(np.mean(x_train[finite_x_train]))
    x_std = float(np.std(x_train[finite_x_train], ddof=0))

    test_mask = np.isfinite(x_test)
    if not test_mask.any():
        return y_pred

    # If the training gene has zero variance, z-scores are undefined.
    # In that case the safest train-only protein-scale prediction is the
    # training protein mean for all finite test RNA values.
    if not np.isfinite(x_std) or x_std <= 0:
        y_pred[test_mask] = y_mean
        return y_pred

    # If the protein has zero variance, this naturally becomes a constant
    # protein-mean prediction.
    if not np.isfinite(y_std):
        y_std = 0.0

    gene_z = (x_test[test_mask] - x_mean) / x_std
    y_pred[test_mask] = y_mean + y_std * gene_z
    return y_pred


# Compatibility alias for any notebook code that still calls the older helper name.
linear_scale_prediction = zscore_gene_to_protein_scale


def build_same_gene_proxy_all_from_lookup(corr_lookup, proxy_type):
    """
    Build all same-name gene/protein proxy rows using notebook-local
    train-only z-normalization from gene scale to protein scale.

    The full-sample same-gene Pearson/Spearman lookup is kept only as metadata
    and for proxy-correlation eligibility analyses. It is not used to compute
    the test prediction normalization.
    """
    rows = []
    possible = sorted(set(Y.columns.astype(str)).intersection(set(X.columns.astype(str))))

    print(f"\nBuilding same-gene {proxy_type} proxy with notebook-local train z-normalization")
    print("- Same-name protein/gene pairs possible:", len(possible))
    print("- Correlation lookup entries:", len(corr_lookup))
    print("- Test samples used:", len(test_samples))
    print("- Train samples used:", len(train_samples))

    for protein in possible:
        c = corr_lookup.get(protein, np.nan)
        if not np.isfinite(c):
            continue

        y_true = Y.loc[test_samples, protein].to_numpy(dtype=float)
        x_test = X.loc[test_samples, protein].to_numpy(dtype=float)

        if USE_TRAIN_ZSCORE_PROXY_NORMALIZATION:
            x_train = X.loc[train_samples, protein].to_numpy(dtype=float)
            y_train = Y.loc[train_samples, protein].to_numpy(dtype=float)
            y_pred = zscore_gene_to_protein_scale(x_train, y_train, x_test)
            variant = "train_zscore_gene_to_protein_scale"
        else:
            y_pred = x_test
            variant = "raw_gene_value"

        metrics = compute_metrics(y_true, y_pred)

        rows.append({
            "protein": protein,
            "run_label": f"proxy_same_gene_{proxy_type}_train_zscore",
            "model_family": "same_gene_proxy",
            "family_group": "proxy",
            "source_folder": f"proxy_same_gene_{proxy_type}",
            "normalization": variant,
            "h5_path": "",
            "proxy_type": proxy_type,
            "same_gene_corr_full": float(c),
            "test_mse": metrics["test_mse"],
            "test_pearson_corr": metrics["test_pearson_corr"],
            "test_cosine_similarity": metrics["test_cosine_similarity"],
            "n_test_used": metrics.get("n_test_used", np.nan),
            "selected_feature_count": 1,
            "train_n": len(train_samples),
            "test_n": len(test_samples),
            "split_hash": globals().get("model_split_hash", ""),
            "best_params": {
                "baseline": "same_name_gene_train_zscore_proxy",
                "proxy_type": proxy_type,
                "normalization": "gene_zscore_to_train_protein_scale",
                "formula": "y_pred = mean(y_train) + std(y_train) * ((x_test - mean(x_train)) / std(x_train))",
                "calibration_fit_samples": "train_only",
                "test_split_source": "read_from_xgb_enet_h5",
            },
        })

    return pd.DataFrame(rows)


def procedural_cosine_cascade(threshold, universe=None):
    """
    Procedural cascade for cosine similarity, matching the Pearson/MSE logic:
    proxy first, then ENet rescue, then XGB rescue, then remaining failures.
    """
    universe = pd.Index(protein_universe if universe is None else universe).astype(str)
    remaining = set(universe)
    rows = []

    proxy_idx = _notebook_index_by_protein(proxy_best_cosine)
    enet_idx = _notebook_index_by_protein(enet_best_cosine)
    xgb_idx = _notebook_index_by_protein(xgb_best_cosine)

    for protein in list(remaining):
        if protein in proxy_idx.index:
            row = proxy_idx.loc[protein]
            value = row.get("test_cosine_similarity", np.nan)
            if np.isfinite(value) and value >= threshold:
                rows.append({
                    "protein": protein,
                    "threshold": threshold,
                    "stage": _notebook_proxy_label(row),
                    "metric": value,
                    "run_label": row.get("run_label", ""),
                })
                remaining.remove(protein)

    for protein in list(remaining):
        if protein in enet_idx.index:
            row = enet_idx.loc[protein]
            value = row.get("test_cosine_similarity", np.nan)
            if np.isfinite(value) and value >= threshold:
                rows.append({
                    "protein": protein,
                    "threshold": threshold,
                    "stage": "ENet",
                    "metric": value,
                    "run_label": row.get("run_label", ""),
                })
                remaining.remove(protein)

    for protein in list(remaining):
        if protein in xgb_idx.index:
            row = xgb_idx.loc[protein]
            value = row.get("test_cosine_similarity", np.nan)
            if np.isfinite(value) and value >= threshold:
                rows.append({
                    "protein": protein,
                    "threshold": threshold,
                    "stage": "XGB",
                    "metric": value,
                    "run_label": row.get("run_label", ""),
                })
                remaining.remove(protein)

    for protein in sorted(remaining):
        rows.append({
            "protein": protein,
            "threshold": threshold,
            "stage": "Still not good",
            "metric": np.nan,
            "run_label": "",
        })

    return pd.DataFrame(rows)


# Synchronize the notebook-local overrides into the utility module's runtime
# globals only, so utility functions that rely on globals can see them during
# this notebook execution. This does not write to utils_ccle.py.
try:
    uc.set_module_globals(globals())
except NameError:
    pass

print("Notebook-local proxy helpers active:")
print("- zscore_gene_to_protein_scale -> train-only gene z-score mapped to protein mean/std")
print("- build_same_gene_proxy_all_from_lookup -> notebook-local override")
print("- procedural_cosine_cascade -> notebook-local helper")
````

## 5. Build train-z-normalized same-gene Pearson and Spearman proxy models

This section builds **all available same-gene proxy predictions first** and stores the full same-gene correlation as metadata. Thresholds are applied later during plotting.

Unlike the raw-proxy version, each proxy prediction maps the test gene value onto the training protein scale using only train-set means and standard deviations:

```python
gene_z = (gene_test - mean(gene_train)) / std(gene_train)
predicted_protein = mean(protein_train) + std(protein_train) * gene_z
```

This means the proxy still uses exactly one RNA feature, but it is normalized to the protein distribution without fitting a regression slope.

## 6. Best-per-family tables

For the procedural cascade, each family contributes its best result per protein:

- Proxy: best of the train-z-normalized same-gene Pearson and same-gene Spearman proxy.
- ENet: best ENet result per protein.

## 15. Why train-z-normalized proxy models should improve cosine similarity and MSE

The distribution plot you provided is consistent with the metric behavior in the raw-proxy notebook: transcriptomics values are mostly non-negative and have a large spike near zero, while proteomics values are centered around zero and include negative values.

That mismatch hurts raw proxy predictions because raw RNA values are compared directly against centered protein values. In this proxy-calibrated notebook, the same-gene proxy is instead fitted as a one-feature linear model on training samples only. This gives the proxy an intercept and slope in protein space before evaluating it on held-out samples.


````python
# Diagnostics for distribution, calibration, and proxy-vs-protein metric behavior.
# These values are computed on the matrices actually loaded by this notebook.

def _distribution_diagnostics(name, df):
    values = df.to_numpy(dtype=float)
    finite = np.isfinite(values)
    vals = values[finite]
    return {
        "matrix": name,
        "n_values": int(vals.size),
        "mean": float(np.mean(vals)) if vals.size else np.nan,
        "std": float(np.std(vals)) if vals.size else np.nan,
        "min": float(np.min(vals)) if vals.size else np.nan,
        "median": float(np.median(vals)) if vals.size else np.nan,
        "max": float(np.max(vals)) if vals.size else np.nan,
        "fraction_negative": float(np.mean(vals < 0)) if vals.size else np.nan,
        "fraction_zero_or_near_zero": float(np.mean(np.isclose(vals, 0.0, atol=1e-12))) if vals.size else np.nan,
    }

xy_distribution_diagnostics = pd.DataFrame([
    _distribution_diagnostics("Transcriptomics X", X),
    _distribution_diagnostics("Proteomics Y", Y),
])

display(xy_distribution_diagnostics)

proxy_metric_diagnostics = (
    proxy_results
    .groupby(["proxy_type", "normalization"], dropna=False)
    .agg(
        proteins=("protein", "nunique"),
        median_proxy_corr_full=("same_gene_corr_full", "median"),
        median_test_pearson=("test_pearson_corr", "median"),
        median_test_cosine=("test_cosine_similarity", "median"),
        median_test_mse=("test_mse", "median"),
        mean_test_mse=("test_mse", "mean"),
    )
    .reset_index()
)

display(proxy_metric_diagnostics)
````

**Interpretation.** Pearson correlation and Spearman correlation mainly measure whether two vectors co-vary across samples. They do not require the two vectors to have the same offset or scale. Spearman is even more forgiving because it only uses ranks.

MSE and cosine similarity ask different questions. MSE compares values directly, so raw non-negative RNA values can have large error when the protein vector is centered around zero or includes negative values. Cosine similarity compares the angle between the raw prediction vector and the protein vector; it is sensitive to centering and sign patterns.

The train-z-normalized proxy fixes the simplest version of that mismatch by mapping RNA values onto the training protein distribution:

```text
gene_z = (gene_test - mean(gene_train)) / std(gene_train)
predicted_protein = mean(protein_train) + std(protein_train) * gene_z
```

This is not a one-gene linear regression fit: the slope is determined only by the ratio of train-set standard deviations, and the intercept is determined only by train-set means. It should improve MSE and often cosine similarity because predictions are now on the protein scale. Pearson/Spearman should usually stay similar to the raw proxy because this is a monotonic linear rescaling when the training standard deviations are positive.

This normalization is not leakage because all means and standard deviations are computed only from `X_train` and `Y_train`. The test protein values are used only for final metric evaluation.

## 18. scTranslator rescue under cosine similarity and MSE

The Pearson rescue section asks whether scTranslator reaches the same Pearson threshold among proteins still not resolved by the base cascade. This added section repeats the same final-rescue theme for the other two metrics:

- **Cosine similarity:** base failures are proteins with cosine similarity below the threshold after proxy -> ENet -> XGB. A scTranslator run rescues a protein when `test_cosine_similarity >= threshold`.
- **MSE:** base failures are proteins with MSE above the threshold after proxy -> ENet -> XGB. A scTranslator run rescues a protein when `test_mse <= threshold`.

For each metric, the notebook saves both the full cascade with scTranslator as a final stage and the focused rescue-only table among base failures when `SAVE_OUTPUTS = True`.

````python
# ---------------------------------------------------------------------
# scTranslator final rescue for cosine similarity and MSE
# ---------------------------------------------------------------------

SC_RESCUE_ORDER = [
    "scTranslator 2M",
    "scTranslator 160K",
    "scTranslator 10K",
    "Still not good after scTranslator",
]


def _sctranslator_stage_from_row(row):
    """Return a stable plot/export stage label for a best scTranslator row."""
    sc_size = row.get("sc_size", np.nan)
    if pd.notna(sc_size) and str(sc_size).strip():
        label = str(sc_size).replace("scTranslator ", "").strip()
    else:
        label = str(row.get("run_label", "")).replace("scTranslator ", "").strip()

    if label in {"2M", "160K", "10K"}:
        return f"scTranslator {label}"
    if label:
        return f"scTranslator {label}"
    return "scTranslator"


def _metric_is_good(value, threshold, higher=True):
    if not np.isfinite(value):
        return False
    return bool(value >= threshold) if higher else bool(value <= threshold)


def _build_sctranslator_rescue_for_metric(
    *,
    metric_name,
    metric_col,
    thresholds,
    base_cascade_func,
    sctranslator_best,
    higher,
):
    """Build full and focused final-rescue assignment tables for one metric."""
    if sctranslator_best is None or len(sctranslator_best) == 0:
        sc_best_idx = pd.DataFrame()
    else:
        sc_best_idx = _index_by_protein(sctranslator_best)

    full_rows = []
    focused_rows = []

    for thr in thresholds:
        base = base_cascade_func(thr).copy()
        base["metric"] = metric_name

        for _, base_row in base.iterrows():
            protein = str(base_row["protein"])
            base_stage = str(base_row["stage"])

            if base_stage != "Still not good":
                out = base_row.to_dict()
                out["threshold"] = thr
                out["metric"] = metric_name
                full_rows.append(out)
                continue

            if len(sc_best_idx) > 0 and protein in sc_best_idx.index:
                sc_row = sc_best_idx.loc[protein]
                value = sc_row.get(metric_col, np.nan)
                rescued = _metric_is_good(value, thr, higher=higher)
                stage = _sctranslator_stage_from_row(sc_row) if rescued else "Still not good after scTranslator"
                run_label = sc_row.get("run_label", np.nan)
                sc_size = sc_row.get("sc_size", np.nan)
            else:
                value = np.nan
                stage = "Still not good after scTranslator"
                run_label = np.nan
                sc_size = np.nan

            row = {
                "protein": protein,
                "threshold": thr,
                "metric": metric_name,
                "stage": stage,
                metric_col: value,
                "run_label": run_label,
                "sc_size": sc_size,
            }
            full_rows.append(row)
            focused_rows.append(row.copy())

    assignments = pd.DataFrame(full_rows)
    rescue_only = pd.DataFrame(focused_rows)
    return assignments, rescue_only


def _show_sctranslator_rescue_metric(
    *,
    metric_name,
    metric_col,
    thresholds,
    base_cascade_func,
    sctranslator_best,
    higher,
    good_label,
    file_prefix,
):
    assignments, rescue_only = _build_sctranslator_rescue_for_metric(
        metric_name=metric_name,
        metric_col=metric_col,
        thresholds=thresholds,
        base_cascade_func=base_cascade_func,
        sctranslator_best=sctranslator_best,
        higher=higher,
    )

    summary = summarize_assignments(assignments, order=CASCADE_ORDER_WITH_SC)
    rescue_summary = summarize_assignments(rescue_only, order=SC_RESCUE_ORDER)

    print(f"Full procedural {metric_name} cascade with scTranslator as final rescue")
    display(
        summary
        .pivot_table(index="threshold", columns="stage", values="proteins", fill_value=0, aggfunc="sum")
        .reindex(columns=CASCADE_ORDER_WITH_SC)
    )

    plot_stacked_partition_rows(
        summary,
        title=f"Procedural {metric_name} cascade with scTranslator final rescue",
        order=CASCADE_ORDER_WITH_SC,
        save_name=f"{file_prefix}_cascade_with_sctranslator.png",
        row_label_func=lambda thr: f"{good_label} {thr:.2f}",
    )

    print(f"Focused scTranslator rescue among proxy/ENet/XGB {metric_name} failures")
    display(
        rescue_summary
        .pivot_table(index="threshold", columns="stage", values="proteins", fill_value=0, aggfunc="sum")
        .reindex(columns=SC_RESCUE_ORDER)
    )

    plot_stacked_partition_rows(
        rescue_summary,
        title=f"Among proxy/ENet/XGB {metric_name} failures: does scTranslator rescue?",
        order=SC_RESCUE_ORDER,
        save_name=f"sctranslator_{file_prefix}_rescue_after_failures.png",
        row_label_func=lambda thr: f"{good_label} {thr:.2f}",
    )

    return assignments, summary, rescue_only, rescue_summary


if len(sctranslator_results) > 0:
    sctranslator_best_cosine = best_per_protein(
        sctranslator_results,
        "test_cosine_similarity",
        higher=True,
    )
    sctranslator_best_mse = best_per_protein(
        sctranslator_results,
        "test_mse",
        higher=False,
    )

    sc_cosine_assignments, sc_cosine_summary, sc_cosine_rescue_only, sc_cosine_rescue_summary = _show_sctranslator_rescue_metric(
        metric_name="cosine similarity",
        metric_col="test_cosine_similarity",
        thresholds=COSINE_GOOD_THRESHOLDS,
        base_cascade_func=procedural_cosine_cascade,
        sctranslator_best=sctranslator_best_cosine,
        higher=True,
        good_label="Cosine >=",
        file_prefix="cosine",
    )

    sc_mse_assignments, sc_mse_summary, sc_mse_rescue_only, sc_mse_rescue_summary = _show_sctranslator_rescue_metric(
        metric_name="MSE",
        metric_col="test_mse",
        thresholds=MSE_GOOD_THRESHOLDS,
        base_cascade_func=procedural_mse_cascade,
        sctranslator_best=sctranslator_best_mse,
        higher=False,
        good_label="MSE <=",
        file_prefix="mse",
    )

    if SAVE_OUTPUTS:
        sc_cosine_assignments.to_csv(EXPORT_TABLE_DIR / "cosine_cascade_with_sctranslator_assignments.csv", index=False)
        sc_cosine_summary.to_csv(EXPORT_TABLE_DIR / "cosine_cascade_with_sctranslator_summary.csv", index=False)
        sc_cosine_rescue_only.to_csv(EXPORT_TABLE_DIR / "sctranslator_cosine_rescue_only_assignments.csv", index=False)
        sc_cosine_rescue_summary.to_csv(EXPORT_TABLE_DIR / "sctranslator_cosine_rescue_only_summary.csv", index=False)

        sc_mse_assignments.to_csv(EXPORT_TABLE_DIR / "mse_cascade_with_sctranslator_assignments.csv", index=False)
        sc_mse_summary.to_csv(EXPORT_TABLE_DIR / "mse_cascade_with_sctranslator_summary.csv", index=False)
        sc_mse_rescue_only.to_csv(EXPORT_TABLE_DIR / "sctranslator_mse_rescue_only_assignments.csv", index=False)
        sc_mse_rescue_summary.to_csv(EXPORT_TABLE_DIR / "sctranslator_mse_rescue_only_summary.csv", index=False)
else:
    print("Skipped: no scTranslator predictions were loaded.")

# Keep moved utility functions synchronized with notebook variables.
try:
    uc.set_module_globals(globals())
except NameError:
    pass
````

````python
# =====================================================================
# Export protein names by cascade/partition threshold and model/stage
#
# Works in:
#   - procedural_ccle.ipynb
#   - 9.procedural_cptac.ipynb
#
# What it saves:
#   1. cascade_partition_proteins_long.csv
#      one protein per row
#
#   2. cascade_partition_proteins_grouped.csv
#      one row per dataset / analysis / threshold / stage,
#      with all protein names joined into one field
#
#   3. cascade_partition_proteins_files_manifest.csv
#      index of all separate files
#
#   4. protein_lists_one_file_per_partition/
#      one separate CSV for each threshold + cascade stage/model
# =====================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re
import hashlib

SAVE_CASCADE_PROTEIN_EXPORTS = True

# ---------------------------------------------------------------------
# Output folders
# ---------------------------------------------------------------------

try:
    EXPORT_TABLE_DIR
except NameError:
    EXPORT_TABLE_DIR = Path("procedural_proxy_z_tables")

EXPORT_TABLE_DIR = Path(EXPORT_TABLE_DIR)
EXPORT_TABLE_DIR.mkdir(parents=True, exist_ok=True)

PROTEIN_EXPORT_DIR = EXPORT_TABLE_DIR / "protein_names_by_cascade_partition"
PROTEIN_EXPORT_DIR.mkdir(parents=True, exist_ok=True)

SPLIT_EXPORT_DIR = PROTEIN_EXPORT_DIR / "protein_lists_one_file_per_partition"
SPLIT_EXPORT_DIR.mkdir(parents=True, exist_ok=True)


# ---------------------------------------------------------------------
# Helpers
# ---------------------------------------------------------------------


# ---------------------------------------------------------------------
# Detect notebook type and collect available assignment tables
# ---------------------------------------------------------------------

is_cptac = (
    _valid_df(_get_global("main_pearson_assignments"))
    or _valid_df(_get_global("pearson_sweep_assignments"))
    or _valid_df(_get_global("mse_percentile_assignments"))
    or _valid_df(_get_global("final_sc_assignments"))
)

if is_cptac:
    default_dataset = "CPTAC"

    candidate_tables = [
        ("main_pearson_cascade", "main_pearson_assignments", _get_global("main_pearson_assignments")),
        ("pearson_threshold_sweep", "pearson_sweep_assignments", _get_global("pearson_sweep_assignments")),
        ("proxy_correlation_threshold_sweep", "proxy_corr_assignments", _get_global("proxy_corr_assignments")),
        ("mse_threshold_sweep", "mse_sweep_assignments", _get_global("mse_sweep_assignments")),
        ("mse_percentile_threshold_sweep", "mse_percentile_assignments", _get_global("mse_percentile_assignments")),
        ("mse_winner_partition", "mse_winner", _get_global("mse_winner")),
        ("mse_procedural_improvement_partition", "mse_improvement", _get_global("mse_improvement")),
        ("cosine_threshold_sweep", "cosine_sweep_assignments", _get_global("cosine_sweep_assignments")),
        ("pearson_cascade_with_sctranslator", "final_sc_assignments", _get_global("final_sc_assignments")),
        ("cosine_cascade_with_sctranslator", "sc_cosine_assignments", _get_global("sc_cosine_assignments")),
        ("sctranslator_cosine_rescue_only", "sc_cosine_rescue_only", _get_global("sc_cosine_rescue_only")),
        ("mse_cascade_with_sctranslator", "sc_mse_assignments", _get_global("sc_mse_assignments")),
        ("sctranslator_mse_rescue_only", "sc_mse_rescue_only", _get_global("sc_mse_rescue_only")),
    ]

else:
    default_dataset = "CCLE"

    candidate_tables = [
        ("main_pearson_cascade", "main_assignments", _get_global("main_assignments")),
        ("pearson_threshold_sweep", "sweep_assignments", _get_global("sweep_assignments")),
        ("main_cosine_cascade", "cosine_main_assignments", _get_global("cosine_main_assignments")),
        ("cosine_threshold_sweep", "cosine_sweep_assignments", _get_global("cosine_sweep_assignments")),
        ("proxy_correlation_threshold_sweep", "proxy_corr_assignments", _get_global("proxy_corr_assignments")),
        ("main_mse_cascade", "main_mse_assignments", _get_global("main_mse_assignments")),
        ("mse_threshold_sweep", "mse_sweep_assignments", _get_global("mse_sweep_assignments")),
        ("pearson_cascade_with_sctranslator", "sc_assignments", _get_global("sc_assignments")),
        ("sctranslator_rescue_only", "sc_rescue_only", _get_global("sc_rescue_only")),
        ("cosine_cascade_with_sctranslator", "sc_cosine_assignments", _get_global("sc_cosine_assignments")),
        ("sctranslator_cosine_rescue_only", "sc_cosine_rescue_only", _get_global("sc_cosine_rescue_only")),
        ("mse_cascade_with_sctranslator", "sc_mse_assignments", _get_global("sc_mse_assignments")),
        ("sctranslator_mse_rescue_only", "sc_mse_rescue_only", _get_global("sc_mse_rescue_only")),
    ]


standardized_tables = []

for analysis_name, source_variable, df in candidate_tables:
    if not _valid_df(df):
        print(f"Skipped missing or empty table: {source_variable}")
        continue

    tmp = _standardize_assignment_table(
        df=df,
        analysis_name=analysis_name,
        source_variable=source_variable,
        default_dataset=default_dataset,
    )

    if _valid_df(tmp):
        standardized_tables.append(tmp)
        print(
            f"Included: {source_variable} -> {analysis_name} "
            f"({len(tmp):,} protein rows)"
        )

if len(standardized_tables) == 0:
    raise ValueError(
        "No assignment tables were found. "
        "Run the cascade/sweep plotting cells first, then rerun this export cell."
    )


# ---------------------------------------------------------------------
# Long protein-level table
# ---------------------------------------------------------------------

cascade_partition_proteins_long = pd.concat(
    standardized_tables,
    ignore_index=True,
)

cascade_partition_proteins_long = (
    cascade_partition_proteins_long
    .sort_values(
        [
            "dataset",
            "analysis",
            "threshold_label",
            "stage",
            "protein",
        ]
    )
    .reset_index(drop=True)
)


# ---------------------------------------------------------------------
# Grouped readable table: one row per graph segment
# ---------------------------------------------------------------------

group_cols = [
    "dataset",
    "analysis",
    "source_variable",
    "threshold_label",
    "stage",
]

cascade_partition_proteins_grouped = (
    cascade_partition_proteins_long
    .groupby(group_cols, dropna=False)
    .agg(
        n_proteins=("protein", "nunique"),
        protein_names=("protein", lambda x: "; ".join(sorted(set(map(str, x))))),
    )
    .reset_index()
    .sort_values(group_cols)
    .reset_index(drop=True)
)

cascade_partition_proteins_grouped["protein_names_preview"] = (
    cascade_partition_proteins_grouped["protein_names"]
    .str.split("; ")
    .apply(lambda xs: "; ".join(xs[:25]) + ("; ..." if len(xs) > 25 else ""))
)


# ---------------------------------------------------------------------
# One file per dataset / analysis / threshold / stage
# ---------------------------------------------------------------------

manifest_rows = []
split_tables = []

for keys, sub in cascade_partition_proteins_long.groupby(group_cols, dropna=False):
    info = dict(zip(group_cols, keys))

    file_stem = "__".join(
        [
            _safe_name(info["dataset"]),
            _safe_name(info["analysis"]),
            _safe_name(info["threshold_label"]),
            _safe_name(info["stage"]),
        ]
    )

    file_name = f"{file_stem}.csv"
    out_path = SPLIT_EXPORT_DIR / file_name

    keep_cols = [
        c for c in [
            "dataset",
            "analysis",
            "source_variable",
            "threshold_label",
            "stage",
            "protein",
            "threshold",
            "prediction_threshold",
            "corr_threshold",
            "mse_percentile_threshold",
            "metric",
            "test_pearson_corr",
            "test_mse",
            "test_cosine_similarity",
            "best_mse",
            "proxy_mse",
            "enet_mse",
            "xgb_mse",
            "same_gene_corr_full",
            "proxy_corr",
            "run_label",
            "model_name",
            "proxy_type",
        ]
        if c in sub.columns
    ]

    split_df = (
        sub[keep_cols]
        .drop_duplicates()
        .sort_values("protein")
        .reset_index(drop=True)
    )

    split_tables.append((out_path, split_df))

    manifest_rows.append(
        {
            "dataset": info["dataset"],
            "analysis": info["analysis"],
            "source_variable": info["source_variable"],
            "threshold_label": info["threshold_label"],
            "stage": info["stage"],
            "n_proteins": split_df["protein"].nunique(),
            "file_name": file_name,
            "relative_path": str(out_path.relative_to(PROTEIN_EXPORT_DIR)),
        }
    )

cascade_partition_proteins_manifest = (
    pd.DataFrame(manifest_rows)
    .sort_values(
        [
            "dataset",
            "analysis",
            "threshold_label",
            "stage",
        ]
    )
    .reset_index(drop=True)
)


# ---------------------------------------------------------------------
# Display BEFORE saving
# ---------------------------------------------------------------------

print("\nGrouped table to be saved:")
display(
    cascade_partition_proteins_grouped[
        [
            "dataset",
            "analysis",
            "source_variable",
            "threshold_label",
            "stage",
            "n_proteins",
            "protein_names_preview",
        ]
    ]
)

print("\nLong protein-level table to be saved, first 100 rows:")
display(cascade_partition_proteins_long.head(100))

print("\nManifest of one-file-per-partition exports to be saved:")
display(cascade_partition_proteins_manifest)


# ---------------------------------------------------------------------
# Save
# ---------------------------------------------------------------------

if SAVE_CASCADE_PROTEIN_EXPORTS:
    long_path = PROTEIN_EXPORT_DIR / "cascade_partition_proteins_long.csv"
    grouped_path = PROTEIN_EXPORT_DIR / "cascade_partition_proteins_grouped.csv"
    manifest_path = PROTEIN_EXPORT_DIR / "cascade_partition_proteins_files_manifest.csv"

    cascade_partition_proteins_long.to_csv(long_path, index=False)
    cascade_partition_proteins_grouped.to_csv(grouped_path, index=False)
    cascade_partition_proteins_manifest.to_csv(manifest_path, index=False)

    for out_path, split_df in split_tables:
        split_df.to_csv(out_path, index=False)

    print("\nSaved protein partition exports:")
    print("Long protein-level table:", long_path)
    print("Grouped protein-name table:", grouped_path)
    print("Manifest:", manifest_path)
    print("One file per threshold/stage:", SPLIT_EXPORT_DIR)
else:
    print("SAVE_CASCADE_PROTEIN_EXPORTS is False, so nothing was saved.")

# Keep moved utility functions synchronized with notebook variables.
try:
    uc.set_module_globals(globals())
except NameError:
    pass
````
